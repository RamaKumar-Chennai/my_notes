````python
Which of the following statements is true about type promotion in PyTorch?

Select an answer

A.
PyTorch always promotes tensors to float32.


B.
Mixed types in a tensor are not allowed.


C.
The higher precision type is selected for the tensor.


D.
Non-numeric data types are allowed in tensors.



The correct answer is:

**C. The higher precision type is selected for the tensor.**

PyTorch follows a **type promotion rule** when combining numeric types. For example:  
- Mixing `int32` and `float32` → the tensor will be promoted to `float32`.  
- Mixing `float32` and `float64` → the tensor will be promoted to `float64`.  

This ensures that computations are consistent and avoid precision loss.  

Key points to remember:  
- **Mixed numeric types are allowed** → PyTorch promotes to the higher precision type.  
- **Non-numeric types (like strings)** are not allowed → that raises an error.  
- PyTorch does **not always default to float32**; it depends on the types involved.  

Would you like me to build a recruiter-style **mini quiz set** on PyTorch type promotion (with formula-based intuition) so you can practice this concept systematically for interviews?
````

````python
Question 13 / 16
What happens if you create a tensor with mixed integers and floats?


Select an answer

A.
An error is raised.


B.
The tensor is promoted to int32.


C.
The tensor is promoted to float32.


D.
The tensor retains mixed data types.


The correct answer is:

**C. The tensor is promoted to float32.**

Here’s the reasoning:  
- PyTorch enforces a single, uniform dtype for all elements in a tensor.  
- If you mix integers and floats, PyTorch applies **type promotion** to avoid precision loss.  
- Since floats are present, the integers are promoted to floats.  
- By default, PyTorch uses **float32** for floating-point tensors unless you explicitly specify another dtype.  

Example:

```python
import torch

tensor = torch.tensor([1.5, 2, 3])
print(tensor)
# tensor([1.5000, 2.0000, 3.0000])
print(tensor.dtype)
# torch.float32
```

👉 So, mixed integers and floats → **float32 tensor**.  

Would you like me to build a **quick recruiter-style MCQ set** contrasting how PyTorch vs NumPy handle mixed types? That’s a common interview trap question.
````

````python
Which type of activation function is most likely to cause the vanishing gradient problem?


Select an answer

A.
ReLU


B.
Sigmoid


C.
Leaky ReLU


D.
Softmax


The correct answer is:

**B. Sigmoid**

Explanation:  
- The **sigmoid activation function** squashes inputs into the range \([0,1]\).  
- For large positive or negative inputs, the gradient becomes extremely small (close to 0).  
- This leads to the **vanishing gradient problem**, where weight updates during backpropagation become negligible, slowing or even halting learning.  
- **ReLU** and **Leaky ReLU** largely avoid this issue because they maintain stronger gradients for positive inputs.  
- **Softmax** is used for output layers in classification, not hidden layers, and does not typically cause vanishing gradients in the same way.  

👉 So, the activation function most likely to cause vanishing gradients is **Sigmoid**.  

Would you like me to also give you a **mini recruiter-style MCQ set** contrasting Sigmoid, Tanh, and ReLU in terms of vanishing/exploding gradients? That’s a common interview trap.
````
