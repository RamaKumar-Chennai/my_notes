````python
Question 8 / 23
What will be the output of the following code?


 


x = torch.tensor(3.0, requires_grad=True)
y = x**2
z = 2 * y
z.backward()
print(x.grad)

Select an answer

A.
12.0


B.
6.0


C.
4.0


D.
None


E.
None of these


A


Let’s walk through the code step by step:

```python
import torch

x = torch.tensor(3.0, requires_grad=True)   # leaf tensor
y = x**2                                    # y = 9
z = 2 * y                                   # z = 18
z.backward()                                # compute gradient
print(x.grad)
```

### Gradient calculation:
- \( y = x^2 \)  
- \( z = 2y = 2x^2 \)  
- So the function is:  
  \[
  z(x) = 2x^2
  \]
- Derivative:  
  \[
  \frac{dz}{dx} = 4x
  \]
- At \( x = 3 \):  
  \[
  \frac{dz}{dx} = 4 \cdot 3 = 12
  \]

### Output:
```text
tensor(12.)
```
````
