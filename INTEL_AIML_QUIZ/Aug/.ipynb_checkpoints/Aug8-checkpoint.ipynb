```python
Question 13 / 13
What is a tool in LangGraph/agent systems?

Select an answer

A.
A pretrained dataset

B.
An external function/API the LLM can use

C.
A GPU kernel

D.
A tokenizer
```

```python
💡 Explanation
In LangGraph or agent-based systems:

A tool is something the language model can call to extend its capabilities beyond text generation.

Examples: a calculator function, a web search API, a database query, or even custom Python code.

The LLM decides when to invoke these tools during reasoning, making it more powerful and interactive.

So, tools are not datasets, GPU kernels, or tokenizers—they’re external functions or APIs that the agent can use to act in the world.
```

```python
Question 12 / 13
What is the purpose of a SystemMessage?

Select an answer

A.
Define model behavior/instructions

B.
Store embeddings

C.
Execute tools

D.
Create checkpoints
```

```python
The correct answer is:

A. Define model behavior/instructions

💡 Explanation
A SystemMessage is used in agent frameworks (like LangChain, LangGraph, or similar systems) to set the rules, tone, and behavior of the language model.

It tells the model how to respond (e.g., “You are a helpful assistant,” “Always answer in JSON,” etc.).

It’s not for storing embeddings, executing tools, or creating checkpoints.

Instead, it acts like the instruction layer that guides the LLM’s behavior throughout the conversation.
```

```python
Question 11 / 13
Which message type usually represents user input?

Select an answer

A.
AIMessage

B.
ToolMessage

C.
HumanMessage

D.
SystemMessage
```

```python
The correct answer is:

C. HumanMessage

💡 Explanation
In agent frameworks like LangChain or LangGraph:

HumanMessage → represents the user’s input (what you type).

AIMessage → represents the model’s response.

SystemMessage → sets rules or instructions for the model’s behavior.

ToolMessage → represents outputs from external tools the model calls.

So whenever you provide input, it’s wrapped as a HumanMessage.
```

```python
Question 10 / 13
What does a node typically return?

Select an answer

A.
GPU tensor only

B.
Updated state values

C.
Name of the next node to execute

D.
Model checkpoint
```

```python
The correct answer is:

B. Updated state values

💡 Explanation
In LangGraph/agent systems, a node is a unit of computation in the graph.

When executed, a node typically returns updated state values (like new messages, variables, or context).

These updated values are then passed along to the next node(s) in the workflow.

It’s not just a GPU tensor, a checkpoint, or the name of the next node—though those can be part of the state. The key idea is that nodes update and 
propagate the state of the agent system.
```

```python
Question 9 / 13
Which method is commonly used to run a LangGraph workflow?

Select an answer

A.
execute()

B.
start()

C.
invoke()

D.
run_graph()
```

```python
The correct answer is:

C. invoke()

💡 Explanation
In LangGraph workflows, the common way to run or execute the graph is by calling the invoke() method.

invoke() takes in the initial input (like a message or state) and runs the workflow through its nodes.

Other names like execute(), start(), or run_graph() are not the standard method in LangGraph.
```

```python
Question 8 / 13
What does graph.compile() do?

Select an answer

A.
Trains the LLM

B.
Converts the graph blueprint into an executable workflow

C.
Creates embeddings

D.
Installs dependencies
```

```python
The correct answer is:

B. Converts the graph blueprint into an executable workflow

💡 Explanation
In LangGraph, you first define your workflow as a graph blueprint (nodes, edges, states).

graph.compile() takes that blueprint and turns it into a runnable workflow object.

After compilation, you can run the workflow using methods like invoke().

It does not train the LLM, create embeddings, or install dependencies—it’s purely about making the graph executable.
```

```python
Question 6 / 13
What is a conditional edge used for?

Select an answer

A.
Training LLMs

B.
Dynamic routing based on conditions

C.
Increasing GPU speed

D.
Creating embeddings
```

```python
The correct answer is:

B. Dynamic routing based on conditions

💡 Explanation
In LangGraph, edges connect nodes. A conditional edge is special because it decides which path the workflow should follow based on some condition (like a variable value, a model output, or user input).

Example: If a prediction confidence is high → go to the "Final Answer" node.

If confidence is low → route to a "Tool Call" node for extra help.

So, conditional edges enable dynamic branching in the workflow, making the agent flexible and adaptive.
```

```python
The correct answer is:

C. StateGraph

💡 Explanation
In LangGraph, the most common object used to define and build a graph is the StateGraph.

It lets you specify nodes, edges, and state transitions.

Once defined, you can compile it into an executable workflow with graph.compile().

Other names like GraphBuilder, LangGraphModel, or AgentBuilder are not the standard constructs in LangGraph.
```

```python
Question 4 / 13
In LangGraph, what is a node?

Select an answer

A.
A database table

B.
A Python function performing a task

C.
A GPU process

D.
A tokenizer
```

```python
The correct answer is:

B. A Python function performing a task

💡 Explanation
In LangGraph, a node is the fundamental building block of the workflow.

Each node is typically implemented as a Python function that performs a specific task (e.g., calling an LLM, executing a tool, transforming data).

Nodes receive input (state), process it, and return updated state values.

By connecting nodes with edges, you define the overall workflow logic.

So, a node is not a database table, GPU process, or tokenizer—it’s essentially a function that does work inside the graph.
```

```python
Question 3 / 13
Choose the most accurate description of State in LangGraph.

Select an answer

A.
State is the latest generation from the LLM

B.
State is only used to track errors during execution

C.
State represents the information that flows through your agent application
```

```python
The correct answer is:

C. State represents the information that flows through your agent application

💡 Explanation
In LangGraph, the State is the central data structure that moves through the workflow:

It contains all the information (messages, variables, context) that nodes read, update, and pass along.

Each node modifies or enriches the state, and the updated state flows to the next node.

It’s not just the latest LLM generation or only error tracking—it’s the complete evolving context of your agent application.

So, the most accurate description is that State is the information flowing through the agent system.
```

```python
Question 2 / 13
Which statement best characterizes LangGraph’s approach to agent design?

Select an answer

A.
LangGraph maximizes freedom, allowing LLMs to make all decisions independently

B.
LangGraph provides strong control over execution flow while still leveraging LLM capabilities for decision making
```

```python
The correct answer is:

B. LangGraph provides strong control over execution flow while still leveraging LLM capabilities for decision making

💡 Explanation
LangGraph’s design philosophy is about structured agent workflows:

You define nodes, edges, and state transitions to control how the agent operates.

This gives developers deterministic control over execution flow.

At the same time, the LLM is still used for reasoning, tool selection, or dynamic decisions where flexibility is needed.

So, instead of giving the LLM complete freedom (option A), LangGraph balances control + intelligence, ensuring reliability while still harnessing the model’s capabilities.
```

```python
Question 1 / 13
What is the primary purpose of LangGraph?

Select an answer

A.
A framework to build control flows for applications containing LLMs

B.
A library that provides interfaces to interact with different LLM models

C.
An Agent library for tool calling
```

```python
The correct answer is:

A. A framework to build control flows for applications containing LLMs

💡 Explanation
The primary purpose of LangGraph is to let developers design structured workflows (graphs) that control how large language models (LLMs) interact with tools, data, and users.

You define nodes (tasks), edges (connections), and state (information flow).

This gives you deterministic control over execution while still leveraging the reasoning power of LLMs.

It’s not just an interface library for LLMs or only an agent library for tool calling—it’s a workflow framework for building reliable agent applications.
```
