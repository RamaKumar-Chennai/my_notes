````python
Question 15 / 18
Which sequence correctly represents the Docker container lifecycle?


 

Select an answer

A.
Image → Running → Build → Stop

B.
Image → Created → Running → Exited → Removed

C.
Container → Image → Running → Deleted

D.
Running → Image → Container → Removed
````

````python

The correct answer is:

**B. Image → Created → Running → Exited → Removed**

### Explanation:
- **A. Image → Running → Build → Stop** → Incorrect. "Build" happens before the image exists, and this sequence is out of order.  
- **B. Image → Created → Running → Exited → Removed** → Correct. This is the standard lifecycle:  
  1. **Image**: You start with an image (e.g., `ubuntu`).  
  2. **Created**: A container is created from the image.  
  3. **Running**: The container is started and executes processes.  
  4. **Exited**: The container stops after completing its task or being stopped manually.  
  5. **Removed**: The container is deleted from the system.  
- **C. Container → Image → Running → Deleted** → Wrong. Images exist before containers, not after.  
- **D. Running → Image → Container → Removed** → Wrong order; doesn’t reflect the lifecycle.  

So the lifecycle is: **Image → Created → Running → Exited → Removed**.  

````

````python
The correct answer is:

**B. The container stops.**

### Explanation:
- **A. The image is deleted** → Wrong. Images remain untouched unless you explicitly remove them with `docker rmi`.  
- **B. The container stops** → Correct. Typing `exit` inside the interactive shell ends the `bash` process. Since that was the main process of the container, the container stops.  
- **C. The container is removed** → Not automatically. The container still exists in a stopped state unless you run `docker rm`.  
- **D. Docker Engine stops** → Wrong. The Docker service itself keeps running; only the container stops.  

So when you type `exit` inside an interactive container, you’re essentially shutting down the container’s main process, which causes the container to **stop but remain available** until you remove it.  

👉 A useful follow-up is: if you want the container to be automatically removed once it stops, you can start it with the `--rm` flag, like this:

```bash
docker run -it --rm ubuntu bash
```  

That way, exiting will both stop and remove the container in one step.
````

````python
The correct answer is:

**D. It is a compressed snapshot containing an application and its environment.**

### Explanation:
- **A. It is a running Linux machine** → Wrong. An image is not running; it’s static.  
- **B. It is a virtual machine** → Wrong. Docker images are not full VMs; they’re lightweight and share the host OS kernel.  
- **C. It is a running Docker daemon** → Wrong. The daemon is the background service that manages containers, not the image itself.  
- **D. It is a compressed snapshot containing an application and its environment** → ✅ Correct. A Docker image is a read-only template that includes everything needed to run an application: code, runtime, libraries, environment variables, and configuration files.  

So, a Docker image is essentially a **blueprint** for containers. Containers are the running instances created from these images.  

👉 A useful analogy: think of an image as a recipe (ingredients + instructions), and a container as the actual dish prepared from that recipe.
````
