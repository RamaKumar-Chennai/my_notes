```python
Question 11 / 26
The final VAE loss is:

Select an answer

A.
Reconstruction Loss − KL Loss


B.
Reconstruction Loss × KL Loss


C.
Reconstruction Loss + KL Loss


n a Variational Autoencoder (VAE), the total loss combines two terms:

Reconstruction Loss → Measures how well the decoder rebuilds the input (e.g., pixel-wise difference).

KL Divergence Loss → Regularizes the latent space by pushing the learned distribution closer to a standard Gaussian.

The final objective is their sum, balancing reconstruction accuracy with latent space regularity.
```

```python
Question 4 / 26
Why can't a standard autoencoder easily generate new data?

Select an answer

A.
It has no decoder


B.
Its latent space is often unstructured


C.
It cannot compress data


D.
It does not use neural networks

B



The correct answer is:

**B. Its latent space is often unstructured**

---

### 🎯 Why?
- A **standard autoencoder** compresses data into a latent vector, but this latent space is not regularized.  
- Because the latent space is **unstructured**, sampling random points from it won’t reliably produce meaningful outputs.  
- That’s why autoencoders are great for **reconstruction** but poor at **generation**.  
- VAEs fix this by enforcing structure in the latent space (via KL Divergence), making sampling possible.  

---

### ⚡ Quick Recall Table

| Option | Meaning | Correct? | Reason |
|--------|---------|----------|--------|
| A. No decoder | False | ❌ | Autoencoders do have decoders |
| B. Latent space unstructured | True | ✅ | Core limitation |
| C. Cannot compress data | False | ❌ | Compression is their strength |
| D. No neural networks | False | ❌ | Autoencoders are neural nets |

👉 Shortcut for interviews:  
**AE = compress + reconstruct, but latent space is messy → can’t easily generate.**  
**VAE = adds structure 🎲 → enables generation.**

---

```

```python
Question 3 / 26
What is the role of the decoder?

Select an answer

A.
Compress data


B.
Add randomness


C.
Reconstruct data from the latent vector


D.
Compute gradients



The correct answer is:

**C. Reconstruct data from the latent vector**

---

### 🎯 Why?
- In both **autoencoders** and **VAEs**, the **decoder’s role** is to take the latent representation \(z\) and **rebuild the original input** (or something close to it).  
- The encoder compresses data into latent space, while the decoder expands it back into the data space.  
- Without the decoder, the model would only produce compressed features but no usable reconstruction.  

---

### ⚡ Quick Recall Table

| Option | Meaning | Correct? | Reason |
|--------|---------|----------|--------|
| A. Compress data | Encoder’s job | ❌ | Not decoder’s role |
| B. Add randomness | Sampling step in VAE | ❌ | Not decoder’s role |
| C. Reconstruct data | Decoder’s job | ✅ | Core function |
| D. Compute gradients | Training mechanism | ❌ | Not decoder’s role |

👉 Shortcut for interviews:  
**Encoder = compress 🧠 → Decoder = reconstruct 🔄**  

---

```
