```python
Question 11 / 20
In a pipeline such as
 
prompt | llm | parser
 
which component is executed last?


 

Select an answer

A.
PromptTemplate


B.
LLM


C.
Output Parser


D.
All run in parallel


C


PromptTemplate → formats the input with placeholders.

LLM → generates the response based on the formatted prompt.

Output Parser → is executed last, taking the raw LLM output and converting it into the desired format (plain string, JSON, structured object, etc.).

So the final step in this chain is the Output Parser.
```

```python
Question 15 / 20
Which execution method generally gives the quickest first response to the user?

Select an answer

A.
batch()


B.
invoke()


C.
stream()


D.
format()


The correct answer is:

C. stream()

📌 Explanation
invoke() → runs the chain once and returns the full response only at the end.

batch() → processes multiple inputs together and returns all results at once.

format() → formats a prompt template but does not call the model.

stream() → returns the generated response token by token, so the user sees the output as it’s being produced.

This makes stream() the method that generally gives the quickest first response to the user, since tokens start appearing immediately instead of waiting for the entire completion.
```

```python
Question 17 / 20
What is the purpose of RunnablePassthrough in complex pipelines?

Select an answer

A.
To discard unnecessary inputs


B.
To encrypt the input


C.
To split the input into tokens


D.
To preserve the original input for later stages


The correct answer is:

D. To preserve the original input for later stages

📌 Explanation
RunnablePassthrough is often used in complex pipelines when you want to keep the original input unchanged alongside other transformations.

It ensures that the raw input is still available for later components, even if other branches or steps modify the data.

So in your MCQ:

A. To discard unnecessary inputs ❌ Wrong — it doesn’t discard anything.

B. To encrypt the input ❌ Wrong — no encryption involved.

C. To split the input into tokens ❌ Wrong — tokenization is handled elsewhere.

D. To preserve the original input for later stages ✅ Correct.
```
