```python
Question 11 / 15
Which evaluation metric is most affected by changes in the decision threshold for a classification model?

Select an answer

A.
Accuracy


B.
AUC-ROC


C.
Precision and Recall

The correct answer is:
C. Precision and Recall
Explanation
- Decision threshold (e.g., 0.5 cutoff for probabilities) directly affects how many cases are classified as positive vs. negative.
- Lowering the threshold → more positives predicted → recall increases, precision may drop.
- Raising the threshold → fewer positives predicted → precision increases, recall may drop.
- Accuracy can change too, but it’s less sensitive because it depends on both positives and negatives.
- AUC‑ROC is threshold‑independent; it evaluates performance across all possible thresholds, so it’s not directly affected by changing one cutoff.
👉 So the metrics most affected by changes in the decision threshold are precision and recall.
```

```python
Question 5 / 21
In a highly imbalanced dataset where the majority class is much larger than the minority class, which metric should you generally avoid relying on?


Select an answer

A.
Precision


B.
Recall


C.
F1 score


D.
accuracy



The correct answer is:
D. Accuracy
Explanation
- In a highly imbalanced dataset, accuracy can be very misleading.
- Example: Suppose 95% of cases are negative and only 5% are positive. A model that predicts everything as negative will still achieve 95% accuracy, but it completely fails to detect the minority class.
- Metrics like precision, recall, and F1 score are much more informative in this scenario because they focus on the minority class performance.
Why not the others
- A. Precision → Useful for measuring false positives.
- B. Recall → Useful for measuring false negatives.
- C. F1 score → Balances precision and recall, often preferred in imbalanced datasets.
👉 So when dealing with imbalanced data, avoid relying on accuracy — it hides poor performance on the minority class.
```

```python
Question 10 / 21
The decision boundary of a logistic regression model is:

Select an answer

A.
Always a straight line


B.
Always a curve


C.
Dependent on the number of features


D.
Not affected by the data

The correct answer is:
C. Dependent on the number of features
Explanation
- In logistic regression, the decision boundary is determined by the linear combination of predictors (features).
- For one feature, the boundary is a single threshold point on the axis.
- For two features, the boundary is a straight line.
- For three or more features, the boundary becomes a hyperplane in higher dimensions.
- So the shape of the decision boundary depends on the dimensionality of the feature space, but it is always linear in the feature space.
```

```python
Question 11 / 21
Which of the following statements is true regarding logistic regression and linear regression?

Select an answer

A.
Logistic regression is used for continuous outcomes, while linear regression is for binary outcomes


B.
Both use a linear function for prediction


C.
Linear regression predicts exact values, while logistic regression predicts probabilities


D.
Logistic regression is more accurate than linear regression for all data types


The correct answer is:
C. Linear regression predicts exact values, while logistic regression predicts probabilities
Explanation
- Linear regression → Used for continuous outcomes. It predicts exact numeric values (e.g., house price, temperature).
- Logistic regression → Used for classification problems. It predicts the probability of belonging to a class (usually binary: 0 or 1).
- Logistic regression applies the sigmoid function to map predictions into the range [0,1], making them interpretable as probabilities.
Why not the others
- A. Logistic regression is used for continuous outcomes, while linear regression is for binary outcomes → This is reversed.
- B. Both use a linear function for prediction → Logistic regression uses a linear combination of inputs, but then applies the sigmoid function, so the final output is not linear.
- D. Logistic regression is more accurate than linear regression for all data types → Accuracy depends on the problem and data; no model is universally “more accurate.”
```

```python
Question 16 / 21
Which of the following machine learning algorithms can be used for imputing missing values?

Select an answer

A.
k-NN


B.
Linear Regression


C.
Logistic Regression

The correct answer is:
A. k‑NN
Explanation
- k‑Nearest Neighbors (k‑NN) can be used for imputing missing values because it relies on the similarity between data points.
- When a feature is missing, k‑NN finds the nearest neighbors (based on other available features) and imputes the missing value using either:
- Majority vote (for categorical data), or
- Average/weighted average (for numerical data).
```
