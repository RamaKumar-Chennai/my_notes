```python
Question 10 / 20
Which of the following sequences is correct for a K-Means algorithm?


 


1. Specify the number of clusters
2. Assign cluster centroids randomly
3. Assign each data point to the nearest cluster centroid
4. Re-assign each point to the nearest cluster centroid
5. Re-compute cluster centroids

Select an answer

A.
1, 2, 3, 4, 5


B.
1, 2, 3, 5, 4


C.
2, 1, 3, 4, 5


D.
None of these

A ANSWER


Explanation of the Sequence
The proper flow of the K‑Means algorithm is:
- Specify the number of clusters (K) → Decide how many groups you want.
- Assign cluster centroids randomly → Initialize starting points for each cluster.
- Assign each data point to the nearest cluster centroid → Based on distance (usually Euclidean).
- Re‑assign each point to the nearest cluster centroid → After centroids shift, points may move to new clusters.
- Re‑compute cluster centroids → Update the centroid positions based on the new assignments.
This cycle (steps 3–5) repeats until stopping criteria are met (centroids stabilize, points stop moving, or max iterations reached).
```

```python
What is a key limitation of K-Means clustering?

Select an answer

A.
It works only with categorical data


B.
It requires the data to follow a linear distribution


C.
It can only work on high-dimensional data


D.
It assumes clusters are spherical



The correct answer is:
D. It assumes clusters are spherical

🔎 Explanation
A key limitation of K‑Means clustering is that it assumes:
- Clusters are spherical (or convex) in shape.
- Each cluster has roughly equal size and density.
- Distance (usually Euclidean) is a good measure of similarity.
Because of this assumption, K‑Means struggles with:
- Non‑spherical clusters (e.g., elongated or irregular shapes).
- Clusters of very different sizes or densities.
- Data with categorical variables (since K‑Means is distance‑based).


```
