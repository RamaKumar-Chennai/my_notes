```python
Question 15 / 17
What is the purpose of the causal mask in decoder self-attention?

Select an answer

A.
To mask padding tokens with zero values


B.
To reduce computational cost to O(n) from O(n²)


C.
To prevent token i from attending to future positions j > i, preserving the autoregressive property


D.
To restrict each head to a local window of tokens



The correct answer is:

**C. To prevent token i from attending to future positions j > i, preserving the autoregressive property** ✅  

---

### 🔎 Explanation
- In **decoder self-attention**, we want the model to generate tokens one by one in an autoregressive fashion.  
- The **causal mask** ensures that when predicting token \(i\), the model only attends to tokens at positions \(\leq i\).  
- This prevents "cheating" by looking ahead at future tokens that haven’t been generated yet.  

---

### ⚡ Quick Recall
- **Padding mask** → hides padding tokens (A is wrong).  
- **Causal mask** → hides *future* tokens (C is correct).  
- Computational cost remains \(O(n^2)\) (B is wrong).  
- Local window restriction is a different technique (like sliding window attention), not causal masking (D is wrong).  

👉 Shortcut for interviews:  
**Causal mask = autoregressive guardrail → no peeking ahead.**

```

```python
Question 14 / 17
 In self-attention, what are the dimensions of the raw score matrix before softmax?

Select an answer

A.
(n × d_model)


B.
(n × n) where n = sequence length


C.
(d_k × d_v)


D.
(batch × d_model)

The correct answer is:

**B. (n × n) where n = sequence length** ✅  

---

### 🔎 Explanation
- In **self-attention**, the raw score matrix is computed as:  
  \[
  \text{Scores} = QK^T
  \]  
- If the sequence length is \(n\), then both \(Q\) and \(K\) have shape \((n, d_k)\).  
- Multiplying \(Q (n \times d_k)\) with \(K^T (d_k \times n)\) gives:  
  \[
  \text{Scores} \in \mathbb{R}^{n \times n}
  \]  
- This matrix represents how **each token attends to every other token** in the sequence before applying softmax.

---

### ⚡ Quick Recall
- **Raw score matrix shape** → \((n \times n)\)  
- **Softmax applied row-wise** → converts scores into attention weights per token.  
- **Output after weighting with V** → contextualized embeddings.  

👉 Shortcut for interviews:  
**Self-attention scores = square matrix (n × n), each row = "who I attend to".**


```

```python
Question 13 / 17
Why do we divide attention scores by √d_k?

Select an answer

A.
To make the output range between 0 and 1


B.
To reduce the number of parameters


C.
To prevent dot products from growing large and pushing softmax into saturation regions with near-zero gradients


D.
To normalize the weight matrices W_Q and W_K


The correct answer is:

**C. To prevent dot products from growing large and pushing softmax into saturation regions with near-zero gradients** ✅  

---

### 🔎 Explanation
- In attention, we compute scores as \(QK^T\).  
- If the dimension \(d_k\) is large, the dot products can grow very large in magnitude.  
- Large values fed into **softmax** cause it to saturate (outputs close to 0 or 1), which leads to **tiny gradients** and slows learning.  
- Dividing by \(\sqrt{d_k}\) keeps the variance of the scores stable, ensuring smoother gradients and more effective training.

---

### ⚡ Quick Recall
- **Without scaling** → unstable training, vanishing gradients.  
- **With scaling by \(\sqrt{d_k}\)** → balanced scores, stable softmax, healthy gradients.  

👉 Shortcut for interviews:  
**Scaling = gradient stabilizer → prevents softmax saturation.**

```

```python
Question 12 / 17
What is the role of softmax in the attention mechanism?

Select an answer

A.
To compute the weighted sum of values


B.
To update the positional encodings


C.
To reduce the dimensionality of embeddings


D.
To normalize attention scores into probabilities


The correct answer is:

**D. To normalize attention scores into probabilities** ✅  

---

### 🔎 Explanation
- In the attention mechanism, we first compute raw scores:  
  \[
  \text{Scores} = \frac{QK^T}{\sqrt{d_k}}
  \]  
- These scores can be any real values (positive or negative).  
- **Softmax** converts them into a probability distribution across tokens:  
  - Each row sums to 1.  
  - Higher scores → higher probability (stronger attention).  
  - Lower scores → lower probability (weaker attention).  
- This ensures the model produces **weighted averages of values (V)** based on meaningful probabilities.

---

### ⚡ Quick Recall
- **Softmax role** → turns raw scores into probabilities.  
- **Probabilities** → used to weight values (V).  
- **Output** → contextualized representation of each token.  

👉 Shortcut for interviews:  
**Softmax = score normalizer → probabilities for attention weights.**

```

```python
Question 4 / 17
How does attention help solve the problem of long input sequences in seq2seq models?

Select an answer

A.
By reducing the input sequence length


B.
By allowing the decoder to focus only on relevant parts of the sequence


C.
By increasing the number of decoder layers


D.
By using a larger batch size


The correct answer is:

**B. By allowing the decoder to focus only on relevant parts of the sequence** ✅  

---

### 🔎 Explanation
- In traditional seq2seq models, the decoder had to rely on a **single fixed-length context vector** from the encoder, which made handling long sequences difficult.  
- **Attention** solves this by letting the decoder dynamically look at different parts of the input sequence at each step.  
- Instead of compressing all information into one vector, the decoder can **focus on the most relevant tokens** for generating the current output.  

---

### ⚡ Quick Recall
- **Problem** → Long sequences lose information when compressed.  
- **Solution** → Attention provides direct access to encoder outputs.  
- **Benefit** → Decoder attends to relevant parts, improving translation, summarization, and other seq2seq tasks.  

👉 Shortcut for interviews:  
**Attention = “smart spotlight” → decoder looks only where it matters.**

```

```python
Question 2 / 17
Which of the following best describes how the attention mechanism works?

Select an answer

A.
 It compresses all input information into a fixed-size context vector.


B.
It directly copies the input sequence to the output.


C.
It replaces the RNN cells in the decoder.


D.
It calculates a weighted sum of encoder hidden states to generate a context vector.



The correct answer is:

**D. It calculates a weighted sum of encoder hidden states to generate a context vector** ✅  

---

### 🔎 Explanation
- The **attention mechanism** doesn’t compress everything into a single fixed vector (A is wrong).  
- Instead, it computes **similarity scores** between the decoder’s current state (Query) and each encoder hidden state (Keys).  
- These scores are normalized with **softmax** to produce **weights**.  
- The weights are then applied to the encoder hidden states (Values), producing a **weighted sum** → the **context vector**.  
- This context vector dynamically changes at each decoding step, letting the model focus on the most relevant parts of the input sequence.  

---

### ⚡ Quick Recall
- **Attention = weighted sum of encoder states**  
- **Weights = importance scores (via Q·K + softmax)**  
- **Context vector = spotlighted information for the decoder**  

👉 Shortcut for interviews:  
**Attention = “smart spotlight” → context vector built from weighted encoder states.**

```
