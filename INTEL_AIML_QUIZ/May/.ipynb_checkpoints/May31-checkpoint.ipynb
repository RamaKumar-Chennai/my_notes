```python
Question 10 / 12
Why are Transformers efficient compared to RNNs for long sequences?

Select an answer

A.
Transformers use convolutional layers for sequence processing.


B.
Transformers discard tokens from earlier positions in a sequence.


C.    
Transformers process all tokens simultaneously using self-attention.


D.
Transformers rely on a single attention mechanism.


The correct answer is:

**C. Transformers process all tokens simultaneously using self-attention** ✅  

---

### 🔎 Why?
- **RNNs** → Process tokens sequentially, step by step. This makes them slow for long sequences and prone to vanishing gradients.  
- **Transformers** → Use **self-attention**, which allows them to look at all tokens in parallel and capture relationships between any two positions in the sequence.  
- This parallelism makes Transformers **much more efficient** for long sequences compared to RNNs.  

👉 Shortcut for interviews:  
**RNN = sequential bottleneck ⏳**  
**Transformer = parallel self-attention ⚡**  


    
```

```python
Question 9 / 12
What is the role of the decoder in the Transformer architecture?

Select an answer

A.
To preprocess input text


B.
To encode text into latent embeddings


C.
To generate output sequences from encoded representations


D.
To improve positional encoding


The correct answer is:

**C. To generate output sequences from encoded representations** ✅  

---

### 🔎 Explanation
- In the **Transformer architecture**, the **encoder** processes the input sequence into a set of contextual embeddings.  
- The **decoder** then takes these embeddings (plus previously generated tokens) and produces the **output sequence** step by step.  
- It uses **self-attention** (with causal masking to preserve autoregression) and **encoder–decoder attention** (to focus on relevant input tokens).  
- This makes the decoder the **generation engine** of the Transformer.  

👉 Shortcut for interviews:  
**Encoder = “understand input” 🧠**  
**Decoder = “generate output” ✍️**  

```

```python
Question 8 / 12
Which of the following is true about the encoder in the Transformer architecture?

Select an answer

A.
It converts the input sequence into a context-aware representation.


B.
It generates a sequence of probabilities for the next word.


C.
It uses convolutional layers to process input.


D.
It directly generates the final output sequence.


The correct answer is:

**A. It converts the input sequence into a context-aware representation** ✅  

---

### 🔎 Explanation
- The **encoder** in the Transformer takes the raw input tokens (after embedding + positional encoding) and processes them through multiple layers of **self-attention** and **feed-forward networks**.  
- This produces **context-aware embeddings**, meaning each token’s representation is enriched with information about its relationship to other tokens in the sequence.  
- These embeddings are then passed to the **decoder**, which uses them to generate the output sequence.  

👉 Shortcut for interviews:  
**Encoder = “understand input” 🧠 → context-aware embeddings**  
**Decoder = “generate output” ✍️ → output sequence**  

```

```python
Question 7 / 12
What is the key characteristic of large language models like GPT?

Select an answer

A.
They use recurrent layers for text generation.


B.
They rely on unsupervised pre-training followed by fine-tuning.


C.
They are specialized for specific domains only.


D.
They are not scalable to large datasets.





The correct answer is:

**B. They rely on unsupervised pre-training followed by fine-tuning** ✅  

---

### 🔎 Explanation
- **Large Language Models (LLMs)** like GPT are trained in two main stages:  
  1. **Unsupervised pre-training** → The model learns general language patterns by predicting the next word across massive text corpora.  
  2. **Fine-tuning** → The model is adapted to specific tasks or domains using supervised data (e.g., Q&A, summarization, dialogue).  

- This two-step process makes LLMs **scalable** and **general-purpose**, unlike domain-specific models.  
- They do **not** rely on recurrent layers (they use Transformer-based self-attention), and they are explicitly designed to handle **large datasets efficiently**.  

👉 Shortcut for interviews:  
**LLM = Pre-train broadly 🧠 → Fine-tune narrowly 🎯**  

```

```python
uestion 6 / 12
Why do Transformers use multi-head attention?

Select an answer

A.
To process multiple inputs simultaneously


B.
To capture different types of relationships in the sequence


C.
To reduce the training time


D.
To improve positional encoding


The correct answer is:

**B. To capture different types of relationships in the sequence** ✅  

---

### 🔎 Explanation
- **Multi-head attention** splits the model’s dimension into multiple smaller “heads.”  
- Each head learns to focus on **different aspects of relationships**:  
  - One head might capture **local syntax** (e.g., subject–verb).  
  - Another might capture **semantic similarity** (e.g., synonyms).  
  - Others might capture **long-range dependencies** across the sequence.  
- By combining these diverse perspectives, the Transformer builds a **richer, more nuanced representation** than a single attention head could.  

👉 Shortcut for interviews:  
**Multi-head = “many lenses” → richer relationships**  

```

```python
Question 5 / 12
Which component in a Transformer model is responsible for capturing relationships between words in a sequence?

Select an answer

A.
Feed-forward layers


B.
Softmax layer


C.
Attention mechanism


D.
Positional encoding

The correct answer is:

**C. Attention mechanism** ✅  

---

### 🔎 Explanation
- The **attention mechanism** is the heart of the Transformer.  
- It allows each word (token) to **attend to other words** in the sequence, capturing relationships such as context, meaning, and dependencies — regardless of distance.  
- This is what enables Transformers to understand nuanced meanings like "bank" in "river bank" vs. "money bank."  
- Other components (feed-forward layers, softmax, positional encoding) play supporting roles, but **attention is the key driver of relational understanding**.  

👉 Shortcut for interviews:  
**Attention = “relationship builder” → connects words across the sequence.**  

```

```python
Question 4 / 12
In Transformer models, what is the purpose of positional encoding?

Select an answer

A.
To indicate the start and end of a sequence


B.
To identify important words in a sequence


C.
To initialize model weights


D.
To incorporate word order information into the model


The correct answer is:

**D. To incorporate word order information into the model** ✅  

---

### 🔎 Explanation
- Transformers rely on **attention**, which has no inherent sense of sequence order.  
- **Positional encoding** injects information about the position of each token in the sequence, ensuring the model can distinguish between "the cat sat on the mat" vs. "the mat sat on the cat."  
- Common approaches include **sinusoidal encodings** (fixed, mathematical functions) or **learned positional embeddings**.  
- Without positional encoding, the model would treat input as a **bag of words** with no order awareness.  

👉 Shortcut for interviews:  
**Positional encoding = “sequence order signal” → tells attention where each token sits.**  

```

```python
Question 3 / 12
What does the term "self-attention" refer to in Transformers?

Select an answer

A.
Attending to external data


B.
A mechanism that allows a model to focus on different parts of its own input sequence


C.
A way to prioritize the training process over inference


D.
A method for reducing model parameters


The correct answer is:

**B. A mechanism that allows a model to focus on different parts of its own input sequence** ✅  

---

### 🔎 Explanation
- **Self-attention** means each token in a sequence can attend to (look at) other tokens in the **same sequence**.  
- This allows the model to capture **contextual relationships** — for example, understanding that "bank" in "river bank" relates to "river," not "money."  
- It’s the core mechanism that gives Transformers their power to model dependencies across long sequences without relying on recurrence.  

👉 Shortcut for interviews:  
**Self-attention = “within sequence focus” → tokens learn from each other.**  

```

```python
Question 2 / 12
What is a major advantage of self-attention over RNNs?

Select an answer

A.
Processes tokens sequentially


B.
Cannot capture long dependencies


C.
Allows parallel computation across tokens


D.
Requires no training

The correct answer is:

**C. Allows parallel computation across tokens** ✅  

---

### 🔎 Explanation
- **RNNs** → Process tokens **sequentially**, one step at a time. This makes them slow and prone to vanishing gradients for long sequences.  
- **Self-attention in Transformers** → Lets the model look at **all tokens simultaneously**, enabling **parallel computation**.  
- This parallelism makes Transformers far more **efficient and scalable** for long sequences compared to RNNs.  

👉 Shortcut for interviews:  
**RNN = sequential bottleneck ⏳**  
**Transformer = parallel self-attention ⚡**  

```

```python
Question 1 / 12
What is the primary innovation introduced by the Transformer model?

Select an answer

A.
Convolutional layers for faster computation


B.
Recurrence mechanisms for sequence modeling


C.
Attention mechanism for capturing dependencies


D.
Gated Recurrent Units for long-term dependencies


The correct answer is:

**C. Attention mechanism for capturing dependencies** ✅  

---

### 🔎 Explanation
- The **Transformer’s breakthrough innovation** was replacing recurrence and convolution with the **attention mechanism**.  
- Attention allows the model to directly capture **dependencies between any two tokens**, regardless of their distance in the sequence.  
- This solved the bottlenecks of RNNs (sequential processing, vanishing gradients) and CNNs (limited receptive fields).  
- By using **self-attention + positional encoding**, Transformers became highly parallelizable and efficient for long sequences.  

👉 Shortcut for interviews:  
**Transformer = “No recurrence, just attention” ⚡**  

```
