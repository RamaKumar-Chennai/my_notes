```python
Question 13 / 20
A spam detection model was trained when spammers used words like "lottery" and "free money." Over time, spammers switched to subtle phishing language and image-based spam. What type of drift occurred?

Select an answer

A.
Data drift


B.
Model drift


C.
Concept drift


D.
Feature drift

C



Recruiter-style intuition
Data drift → input distribution changes (e.g., more emails from a new region).

Concept drift → the meaning of features changes (e.g., “lottery” no longer signals spam, but image attachments do).

Concept drift is more dangerous because the model’s logic itself becomes invalid.
```

```python
Question 7 / 20
Which tool is commonly used for experiment tracking?
 

Select an answer

A.
TensorFlow Lite


B.
Numpy


C.
Flask


D.
MLflow

D



Explanation
MLflow is a widely used tool for experiment tracking in ML projects.

It allows you to log:

Parameters (learning rate, batch size, etc.)

Metrics (accuracy, loss, etc.)

Artifacts (models, plots, datasets)

Environment details (dependencies, versions)

This makes experiments reproducible and comparable, which is essential in MLOps.

⚖️ Recruiter-style intuition
TensorFlow Lite → for deploying models on mobile/edge devices.

NumPy → numerical computing library, not for experiment tracking.

Flask → lightweight web framework, not ML-specific.

MLflow → purpose-built for experiment tracking and model management.
```
