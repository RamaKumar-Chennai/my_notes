```python
Question 13 / 18
Why are sparse matrices used in machine learning and data science?

Select an answer

A.
They require more memory to store than dense matrices


B.
They are easier to manipulate computationally than dense matrices


C.
They can efficiently store data with many zero elements, saving memory


D.
They provide more accurate results in calculations



The correct answer is:

C. They can efficiently store data with many zero elements, saving memory ✅

Sparse matrices are used in machine learning and data science because most text or high‑dimensional datasets contain a huge number of zeros 
(for example, in Bag of Words or TF‑IDF representations). Instead of storing every zero explicitly, sparse matrix formats only store the non‑zero 
values and their positions, which drastically reduces memory usage and speeds up computation. Dense matrices, by contrast, would waste space and
processing power on all those zeros.
```

```python
Question 14 / 18
What is the typical space complexity advantage of using sparse matrices over dense matrices?

Select an answer

A.
Sparse matrices typically use space proportional to the number of elements.


B.
Sparse matrices typically use space proportional to the number of non-zero elements.


C.
Sparse matrices typically use space proportional to the square of the number of elements.


D.
Sparse matrices typically use space proportional to the logarithm of the number of elements.



    The correct answer is:

B. Sparse matrices typically use space proportional to the number of non‑zero elements ✅

Dense matrices store all entries, including zeros, so their space complexity is proportional to the total number of elements. Sparse matrices, on the
    other hand, only store the non‑zero values and their positions, which makes them far more memory‑efficient when the majority of entries are zero.
    This advantage is especially important in text mining (Bag of Words, TF‑IDF) and recommender systems, where datasets are extremely high‑dimensional 
    but mostly empty.
```

```python
Question 5 / 18
What do vectorizers do?

Select an answer

A.
They convert the entire dataset into a vector


B.
They convert each target label to a vector


C.
They convert each feature into a vector


D.
They convert each sample into a vectoR




D



Vectorizers are tools in text processing that transform raw text into numerical feature vectors so machine learning models can work with them. 
Each document (or sample) is converted into a vector representation, where dimensions correspond to words or features (like counts, TF‑IDF scores, 
                                                                                                                       or hashed indices). 
This allows algorithms to analyze text mathematically, enabling tasks such as classification, clustering, or sentiment analysis.

👉 Shortcut for interviews:

CountVectorizer → raw word counts per sample

TF‑IDF Vectorizer → weighted importance per sample

```
