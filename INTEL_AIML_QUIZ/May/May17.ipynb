````python
Question 18 / 20
What is the default initialization of the hidden state in PyTorch RNNs?

Select an answer

A.
Random values from a uniform distribution


B.
Small random Gaussian values


C.
Pre-trained weights


D.
All zeros


D


The correct answer is:

**D. All zeros**

---

### 🧠 Explanation:
- In PyTorch, when you don’t explicitly provide an initial hidden state to an RNN (`nn.RNN`, `nn.LSTM`, or `nn.GRU`), the framework **defaults to 
initializing it with zeros**.  
- This means both the hidden state \(h_0\) and, for LSTMs, the cell state \(c_0\) start as zero tensors.  
- The shape of these tensors depends on:
  - `num_layers` × `num_directions` (for bidirectional RNNs)  
  - `batch_size`  
  - `hidden_size`  

### Example:
```python
import torch
import torch.nn as nn

rnn = nn.RNN(input_size=10, hidden_size=20, num_layers=2)
x = torch.randn(5, 3, 10)  # (seq_len, batch, input_size)

# No hidden state provided → defaults to zeros
output, h_n = rnn(x)

print(h_n.shape)  # torch.Size([2, 3, 20])
```

- **A. Random values from a uniform distribution** → ❌ not the default.  
- **B. Small random Gaussian values** → ❌ also not default.  
- **C. Pre-trained weights** → ❌ only if you load them manually.  
- **D. All zeros** → ✅ correct default initialization.  

---

````

````python
Question 17 / 20
Which of the following is true about torch.nn.RNNCell and torch.nn.RNN?

Select an answer

A.
Both process the entire sequence at once


B.
RNN processes one time step at a time, while RNNCell processes the entire sequence


C.
RNNCell processes a single time step, while RNN processes the entire sequence


D.
Both are interchangeable for any RNN implementation



The correct answer is:

**C. RNNCell processes a single time step, while RNN processes the entire sequence**

---

### 🧠 Explanation:
- **`torch.nn.RNNCell`**  
  - Represents the computation for **one time step** of an RNN.  
  - You must manually loop over the sequence and feed inputs step by step.  
  - Useful when you want fine-grained control over the recurrence.  

- **`torch.nn.RNN`**  
  - Wraps the recurrence logic to process the **entire sequence at once**.  
  - Handles looping internally, so you just pass the whole sequence tensor.  
  - More convenient for standard use cases.  

### Example:
```python
import torch
import torch.nn as nn

# RNNCell: single step
cell = nn.RNNCell(input_size=10, hidden_size=20)
x = torch.randn(3, 10)   # batch=3, input_size=10
h = torch.zeros(3, 20)   # initial hidden state
h = cell(x, h)           # one time step

# RNN: entire sequence
rnn = nn.RNN(input_size=10, hidden_size=20)
seq = torch.randn(5, 3, 10)  # seq_len=5, batch=3, input_size=10
output, h_n = rnn(seq)       # processes all 5 steps internally
```

- **A. Both process the entire sequence at once** → ❌ incorrect.  
- **B. RNN processes one time step, while RNNCell processes the entire sequence** → ❌ reversed.  
- **C. RNNCell processes a single time step, while RNN processes the entire sequence** → ✅ correct.  
- **D. Both are interchangeable for any RNN implementation** → ❌ not true; they serve different purposes.  

---

````

````python
Question 16 / 20
How can you extract only the final output from a sequence of outputs in PyTorch RNNs?

Select an answer

A.
Use the last index of the output tensor


B.
Use torch.nn.functional.softmax


C.
Use torch.max on the output tensor


D.
Use torch.sum on the hidden states



The correct answer is:

**A. Use the last index of the output tensor**

---

### 🧠 Explanation:
- In PyTorch RNNs (`nn.RNN`, `nn.LSTM`, `nn.GRU`), the forward pass returns:
  1. **`output`** → contains the hidden states for **all time steps** (shape: `(seq_len, batch, hidden_size)`).
  2. **`h_n`** → contains the **final hidden state(s)** (shape: `(num_layers * num_directions, batch, hidden_size)`).

- If you want the **final output from the sequence**, you can simply index the last time step:
```python
import torch
import torch.nn as nn

rnn = nn.RNN(input_size=10, hidden_size=20)
x = torch.randn(5, 3, 10)   # seq_len=5, batch=3, input_size=10
output, h_n = rnn(x)

final_output = output[-1]   # last time step output
print(final_output.shape)   # torch.Size([3, 20])
```

### Why not the other options?
- **B. `torch.nn.functional.softmax`** → ❌ applies normalization, not extraction.  
- **C. `torch.max` on the output tensor** → ❌ gives max values, not the final step.  
- **D. `torch.sum` on the hidden states** → ❌ aggregates values, not the final output.  

---
````

````python
Question 10 / 20
What is the main cause of the vanishing gradient problem in RNNs?

Select an answer

A.
Use of large learning rates


B.
Insufficient training data


C.
Repeated multiplication of small gradients through many time steps


D.
High dimensionality of input data



The correct answer is:

**C. Repeated multiplication of small gradients through many time steps**

---

### 🧠 Explanation:
- In RNNs, gradients are propagated backward through **each time step**.  
- This involves repeated multiplication by weight matrices and activation derivatives.  
- If these values are **less than 1**, the gradient shrinks exponentially as it moves backward → **vanishing gradient problem**.  
- Conversely, if values are **greater than 1**, gradients can explode.  

### Example intuition:
Suppose each step multiplies the gradient by 0.9.  
After 50 steps:  
\[
0.9^{50} \approx 0.005
\]  
The gradient becomes almost zero, making it impossible for the network to learn long-term dependencies.

### Option review:
- **A. Large learning rates** → ❌ can cause instability, but not vanishing gradients.  
- **B. Insufficient training data** → ❌ affects generalization, not gradient flow.  
- **C. Repeated multiplication of small gradients through many time steps** → ✅ correct cause.  
- **D. High dimensionality of input data** → ❌ unrelated to vanishing gradients.  

---

````
