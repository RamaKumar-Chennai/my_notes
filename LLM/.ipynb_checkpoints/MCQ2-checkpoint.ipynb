```python
1. What is NLP?
A subfield of Computer science to understand and process natural language using AI
A search engine
An image editor
None of the above
Answer: A) A subfield of Computer science to understand and process natural language using AI

Explanation:

NLP, Natural Language Processing, is a subfield of Computer Science to understand and process natural language with the help of Artificial 
Intelligence, AI.

Discuss this question

2. What is Ambiguity in the context of NLP?
Ability of being understood in multiple ways
Inability to understand a word
Both of the above
None of the above
Answer: A) Ability of being understood in multiple ways

Explanation:

Natural Language is ambiguous by nature. Many words/ sentences can have different meanings in different contexts. In the case of NLP, this possibility
of being understood is known as Ambiguity.

Discuss this question

3. What is Lexical Ambiguity in the context of NLP?
Ambiguity of a sentence
Ambiguity of single word
Ambiguity of phrase of a sentence
None of the above
Answer: B) Ambiguity of single word

Explanation:

Lexical Ambiguity refers to the ambiguity due to a single word. For example, a word Silver can be treated as a noun, adjective or as a verb.

Discuss this question

4. What is Syntactic Ambiguity in the context of NLP?
Ambiguity of a sentence
Ambiguity of single word
Ambiguity of phrase of a sentence
None of the above
Answer: A) Ambiguity of a sentence

Explanation:

When a sentence can be parsed in multiple ways, the ambiguity caused is termed as Syntactic Ambiguity. Consider the case in the following sentence,
'Man saw a boy with a magnifying glass'. Here the sentence is ambiguous as to whether the man saw a boy carrying a magnifying glass or man saw the boy
with the help of a magnifying glass.

Discuss this question

5. What is Semantic Ambiguity in the context of NLP?
Ambiguity of a sentence
Ambiguity of single word
Ambiguity of phrase of a sentence
None of the above
Answer: C) Ambiguity of phrase of a sentence

Explanation:

When a sentence contains ambiguous words/phrases, the ambiguity caused is termed as Syntactic Ambiguity. Consider the case in the following sentence,
'A truck hit the pole when it was moving'. Here the sentence is ambiguous as to whether the truck, while moving hit the pole or truck hit the moving 
pole.

Discuss this question

6. What is Anaphoric Ambiguity in the context of NLP?
Ambiguity of a sentence
Ambiguity of single word
Ambiguity of phrase of a sentence
None of the above
Answer: D) None of the above

Explanation:

Anaphoric ambiguity arises when anaphoric entities are used. Consider the following sentences. A goat rode up the hill. It was very steep. It reached 
quickly though. Here use of It causes anaphoric ambiguity.

Discuss this question

7. What is Pragmatic Ambiguity in the context of NLP?
Ambiguity of a context of phrase
Ambiguity of single word
Ambiguity of a sentence
None of the above
Answer: A) Ambiguity of a context of phrase

Explanation:

Pragmatic ambiguity arises when the context of a sentence or phrase can be interpreted in many ways. Consider a case of 'I like you too' can be 
interpreted as 'I like you' as you like me or 'I like you' as anyone does.

Discuss this question

8. Which of the following is true about Morphological Processing in the context of NLP?
It is the first phase of NLP
It breaks the chunks of language into tokens
Both of the above
None of the above
Answer: C) Both of the above

Explanation:

Morphological Processing is the first phase of NLP. Main purpose of Morphological processing is to break the chunks of the language to tokens 
corresponding to words, sentences or paragraphs. For example, 'uneasy' can be divided into a subword token as 'un-easy'.

Discuss this question

9. Which of the following is true about Syntax Analysis in the context of NLP?
It is the second phase of NLP
It checks if the sentence is well formed
It creates a structure showing the syntactical relationship between words
All of the above
Answer: D) All of the above

Explanation:

Syntax Analysis is the second phase of NLP. Main purpose of the Syntax Analysis phase is twofold. First to check if the sentence is well formed and 
secondly to create a structure showing the syntactical relationship between words. For example, an invalid sentence like 'The school is going to the boy.' will be rejected by the Syntax Analysis phase.

Discuss this question

10. Which of the following is true about Semantic Analysis in the context of NLP?
It is the third phase of NLP
It checks the meaningfulness of a sentence
Both of the above
None of the above
Answer: C) Both of the above

Explanation:

Semantic Analysis is the third phase of NLP. Main purpose of the Semantic Analysis phase is to check the meaningfulness of a sentence like the 
dictionary meaning. For example, semantic analysis will reject a sentence like 'Hot icecream.'

Discuss this question

11. Which of the following is true about Pragmatic Analysis in the context of NLP?
It is the fourth phase of NLP
It fits the objects/events with references.obtained in semantic analysis
Both of the above
None of the above
Answer: C) Both of the above

Explanation:

Pragmatic Analysis is the fourth and last phase of NLP. Main purpose of the Pragmatic Analysis phase is to set the objects/events obtained in the 
previous step, semantic analysis, at their right place. For example, 'Put the ball in the basket on a shelf' may have two semantic interpretations. 
Pragmatic analysis will choose one of the possibilities.

Discuss this question

12. What is Corpus in terms of NLP?
A large structured set of machine-readable text
Dictionary of words
Both of the above
None of the above
Answer: A) A large structured set of machine-readable text

Explanation:

Corpus is a large structured set of machine-readable text produced during communications. A corpus can be derived from multiple sources like data 
available in electronic form, transcripts or even using OCR, Optical Character Recognition etc.

Discuss this question

13. Which of the following is an element of Corpus Design?
Corpus Balance
Sampling
Corpus Size
All of the above
Answer: D) All of the above

Explanation:

Corpus Balance, Sampling and Corpus Size all are integral parts of a Corpus Design. Corpus balance refers to the wide range of the text categories as representatives of the language. Sampling is used to get a subset of data which can be used to represent the large set of information and Corpus Size represents the size of the Corpus.

Discuss this question

14. Which of the following is a valid type of Corpus?
TreeBank Corpus
PropBank Corpus
Both of the above
None of the above
Answer: C) Both of the above

Explanation:

TreeBank and PropBank, both are types of Corpus. TreeBank Corpus is a linguistic parsed text-based corpus which annotates syntactic or semantic sentence
structure whereas PropBank Corpus or more specifically Proposition Bank Corpus is a corpus which is annotated with verbal propositions and their 
arguments.

Discuss this question

15. Which of the following are valid types of TreeBank Corpus?
Semantic TreeBanks
Syntactic TreeBanks
Both of the above
None of the above
Answer: C) Both of the above

Explanation:

Semantic TreeBanks and Syntactic TreeBanks are the most common types of TreeBanks.Semantic TreeBank uses formal representation of semantic structure of sentences. Syntactic TreeBank uses expressions of formal language retrieved from parsed treebank data.

Discuss this question

16. Which of the following is a Semantic TreeBank Corpus?
Robot Commands TreeBanks
Geoquery
Groningen Meaning Bank
All of the above
Answer: C) Groningen Meaning Bank

Explanation:

Robot Commands Treebank, Geoquery, Groningen Meaning Bank, RoboCup Corpus are popular Semantic Treebanks.

Discuss this question

17. Which of the following is a Syntactic TreeBank Corpus?
Penn Arabic TreeBank
Columbia Arabic TreeBank
Sinina
All of the above
Answer: D) All of the above

Explanation:

Penn Arabic Treebank, Columbia Arabic Treebank are syntactic Treebanks in Arabia language. Sininca is a syntactic Treebank in Chinese language. Lucy, Susane and BLLIP WSJ syntactic treebank corpus in English language.

Discuss this question

18. What is Morphological Parsing in terms of NLP?
Parsing of morphemes
Breaking down words into smaller meaning units called morphemes
Both of the above
None of the above
Answer: C) Both of the above

Explanation:

Morphological parsing refers to parsing of morphemes. A morpheme represents the smallest meaningful units. For example, consider a word, boxes, we can break it into two words/morphemes box and es.

Discuss this question

19. Morphology is a study of?
Formation of words
Origin of words
Grammatical formation of words
All of the above
Answer: D) All of the above

Explanation:

Morphology is study of following:

Words formation.
Words origin.
Grammatical formations of the words.
Usage of prefixes and suffixes in word formation.
Formation of parts-of-speech (PoS) of a language.
Discuss this question

20. Which of the following is a valid type of Morpheme?
Stem
Word Order
Both of the above
None of the above
Answer: C) Both of the above

Explanation:

Morphemes are the smallest meaningful units and can be divided into following types:

Stems
Word Order
Discuss this question

21. Which of the following correctly defines the role of a parser?
A Parser reports any syntax error
A Parser recovers from a common error to continue processing of the rest of the program
A Parser creates a parse tree
All of the above
Answer: D) All of the above

Explanation:

A Parser is a software component which takes text as input and gives structural representation of input after checking it against the grammar of the 
language. Main roles of a parser are following:

To report any syntax error.
To recover from commonly occurring errors to continue parsing of the rest of the program.
To create a parse tree
To create a symbol tree
To produce intermediate representations.
Discuss this question

22. Which of the following correctly defines Top-Down parser?
Top-Down Parser starts parsing the tree from the start symbol, constructing the tree up to the input symbol
Top-Down Parser starts parsing from the input symbol and constructs the tree up to the start symbol
Both of the above
None of the above
Answer: A) Top-Down Parser starts parsing the tree from the start symbol, constructing the tree up to the input symbol

Explanation:

Top-Down Parser starts parsing the tree from the start symbol, constructing the tree up to the input symbol. Top-Down parser uses a recursive procedure to process the inputs.

Discuss this question

23. Which of the following correctly defines Bottom-Up parser?
Bottom-Down Parser starts parsing the tree from the start symbol, constructing the tree up to the input symbol
Bottom-Down Parser starts parsing from the input symbol and constructs the tree up to the start symbol
Both of the above
None of the above
Answer: B) Bottom-Down Parser starts parsing from the input symbol and constructs the tree up to the start symbol

Explanation:

Bottom-Down Parser starts parsing the input symbol and constructs the tree up to the start symbol.

Discuss this question

24. Which of the following correctly defines left-most derivation?
Sentential form of an input is scanned and replaced from left to right
Sentential form of an input is scanned and replaced from right to the left
Both of the above
None of the above
Answer: A) Sentential form of an input is scanned and replaced from left to right

Explanation:

Sentential form or more specifically left sentential form of an input is scanned and replaced from left to the right in left-most derivation

Discuss this question

25. Which of the following correctly defines right-most derivation?
Sentential form of an input is scanned and replaced from left to right
Sentential form of an input is scanned and replaced from right to the left
Both of the above
None of the above
Answer: B) Sentential form of an input is scanned and replaced from right to the left

Explanation:

Sentential form or more specifically right sentential form of an input is scanned and replaced from right to the left in right-most derivation.

Discuss this question

26. Which of the following is correct about Parse Tree?
It is a graphical representation of a derivation
Root of the parse tree is denoted by the start symbol of derivation
Terminals are denoted by leaf nodes and non-terminals are denoted by internal nodes
All of the above
Answer: D) All of the above

Explanation:

A parse tree can be defined as a graphical representation of a derivation. Root of the parse tree denotes the start symbol of derivation. Leaf nodes 
denote the terminals and internal nodes denote the non-terminals.
```

```python
What is the recognized statement of the Maximum Matching algorithm for the string 'thetabledownthere'?
A both (a) and (b)
the table down there
theta bled own there
D None of the above
Explanation: Maximum Matching Algorithm in Word Segmentation
The Maximum Matching algorithm is a greedy approach used in word segmentation, especially for languages without explicit word boundaries (like Chinese or for concatenated strings in NLP tasks). It attempts to match the longest possible word from a dictionary starting from the beginning of the string.
Given the string:
"thetabledownthere"

Two plausible segmentations using Maximum Matching are:
- "the table down there" – all valid English words.
- "theta bled own there" – also composed of valid English words, though less semantically coherent.
Since both segmentations consist of valid dictionary words, both are recognized outputs of the algorithm depending on the dictionary used and tie-breaking rules.

🔍 Why Not Just One?
- Maximum Matching is not deterministic when multiple valid segmentations exist.
- It depends on the dictionary and whether it prioritizes semantic coherence or just longest matches.
- In some implementations, both outputs may be considered valid and returned for further disambiguation.
```

```python
Excellent challenge, Rama 👏 — this is exactly the kind of subtle distinction recruiters love to test. Let’s unpack it carefully.

---

## 🔎 What is a "language model" in NLP?
- Traditionally, a **language model (LM)** is defined as something that learns the probability distribution of sequences of words/tokens.  
- In practice, this means it can **predict the next word** (autoregressive LM like GPT) or **predict missing words** (masked LM like BERT).

---

## 🧩 Why BERT is still called a *pretrained language model*
- **Training objective:**  
  - BERT is trained with **Masked Language Modeling (MLM)** → randomly masks tokens and predicts them from context.  
  - It also uses **Next Sentence Prediction (NSP)** → predicts whether one sentence follows another.  
- These are **language modeling tasks** — even though BERT doesn’t generate fluent text like GPT, it *does* learn the statistical structure of language.
- **Output:**  
  - Instead of generating text, BERT produces **contextual embeddings** (vectors) for each token.  
  - These embeddings encode rich semantic and syntactic information because they were learned via language modeling objectives.

---

## ✅ Recruiter‑friendly phrasing
You can say:  
- *“BERT is a pretrained language model because it was trained on language modeling tasks (masked word prediction and next sentence prediction). Although it doesn’t generate text directly, it learns contextual token embeddings that capture the statistical structure of language, which downstream tasks can use.”*

---

## ⚖️ Side‑by‑side comparison
| Model | Training Objective | Output | Usage |
|-------|-------------------|--------|-------|
| GPT   | Next‑word prediction (autoregressive LM) | Generates text | Text generation, chatbots |
| BERT  | Masked word prediction + NSP (bidirectional LM) | Contextual embeddings | Classification, QA, NER |

---

## 🎯 Key takeaway for interviews
- **Language model ≠ text generator.**  
- A language model is defined by its **training objective on language data**, not necessarily by whether it produces fluent sentences.  
- BERT qualifies because it learns language representations via LM tasks, even though its practical output is embeddings.

---

Would you like me to show you a **numeric demo** where we mask a token in a sentence (e.g., “The cat sat on the [MASK]”) and let BERT predict it, so you can see concretely how it acts as a language model? That would give you copy‑paste‑ready proof for interviews.
```

```python
Great question, Rama — this is one of those subtle distinctions that trips people up in interviews. Let’s break it down clearly.

---

## 🔎 What is Language Modelling?
- **Definition:** Language modelling is the task of learning the probability distribution of sequences of words/tokens in a language.  
- In simple terms: it’s about predicting how likely a word (or sequence of words) is, given the context.  
- This makes the model capable of understanding and generating human‑like text.

---

## 🧩 Types of Language Modelling Tasks
There are different ways to train a language model, depending on the objective:

| Task | Example | Model Type | Purpose |
|------|---------|------------|---------|
| **Next‑word prediction (autoregressive LM)** | “The cat sat on the ___” → predict *mat* | GPT | Text generation |
| **Masked word prediction (MLM)** | “The cat sat on the [MASK]” → predict *mat* | BERT | Contextual embeddings |
| **Next Sentence Prediction (NSP)** | “The cat sat on the mat. [SEP] It was tired.” → Is second sentence logically next? | BERT | 
Sentence‑level understanding |

---

## ✅ Answering your question
Yes — **masked word prediction and NSP are examples of language modelling tasks.**  
- They are *not the only ones*, but they are the specific objectives used to pretrain **BERT**.  
- GPT uses next‑word prediction, while BERT uses masked word prediction + NSP.  
- All of these fall under the umbrella of **language modelling**, because they involve learning the statistical structure of language.

---

## 🎯 Recruiter‑friendly phrasing
You can say:  
- *“Language modelling is the process of learning the probability distribution of words in a sequence. Tasks like masked word prediction and
    next sentence prediction are specific language modelling objectives used to pretrain models like BERT.”*

---

Would you like me to show you a **numeric demo** where we mask a word in a sentence and let BERT predict it, so you can see concretely how masked word prediction works as a language modelling task? That would give you copy‑paste‑ready proof for your notes.
```
