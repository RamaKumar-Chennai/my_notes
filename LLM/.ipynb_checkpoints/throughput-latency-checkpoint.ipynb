```python
Throughput means:

✅ How many requests/tasks a system can process in a given time.

For LLMs:

If an LLM serves 10 users per second, throughput = 10 requests/sec.
If another serves 1,000 users per second, throughput = 1,000 requests/sec.

Higher throughput = more users can be served simultaneously.

Simple example

Suppose 1,000 users are asking questions to your chatbot.

Tool	Requests handled per secondFlask server	50
vLLM	500+
Optimized vLLM cluster	Thousands

vLLM is built specifically to maximize throughput while efficiently using GPU memory.

Assessment Definition

Throughput = Number of requests (or tokens) processed per unit time.

Throughput vs Latency
Latency = How long one user waits for a response.
Throughput = How many total users/requests can be handled.

Example:

Response arrives in 1 second → Low latency ✅
System serves 1000 users/sec → High throughput ✅

For LLM deployment interviews/tests, remember:

Latency = Speed for one request
 Throughput = Capacity for many requests

That's why vLLM is the correct answer for high-throughput LLM serving. ✅
```
