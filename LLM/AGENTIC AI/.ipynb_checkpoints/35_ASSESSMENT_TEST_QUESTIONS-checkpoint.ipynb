```python
Sure, Rama. For assessment preparation, memorize these 35 Agentic AI Questions with concise answers.

1. What is Agentic AI?

Answer:
 Agentic AI refers to AI systems that can autonomously plan, reason, use tools, take actions, observe results, and iteratively work toward a goal.

2. Difference between LLM and Agent?

LLM:

Generates text based on prompt.
No decision-making.
No tool execution.

Agent:

Uses LLM + Tools + Memory + Planning.
Can take actions.
Can execute multi-step tasks.
3. What are the main components of an AI Agent?

Answer:

LLM (Brain)
Memory
Tools
Planning Module
Execution Engine
Observation/Feedback Loop
4. What is the ReAct Framework?

Answer: ReAct = Reason + Act

Flow:

Thought
↓
Action
↓
Observation
↓
Thought
↓
Final Answer


Allows agents to reason before taking action.

5. What is Tool Calling?

Answer:

Tool calling allows an LLM to invoke external functions/APIs.

Example:

User → Weather in Chennai?
LLM → Call Weather API
API → 34°C
LLM → Respond

6. Why are tools important in Agentic AI?

Answer:

LLMs don't know:

Real-time data
Databases
Calculators
APIs

Tools extend capabilities.

7. What is a Planning Agent?

Answer:

A planning agent breaks a large goal into smaller tasks before execution.

Example:

Build application
↓
Gather requirements
↓
Design
↓
Code
↓
Test

8. What is Chain-of-Thought (CoT)?

Answer:

LLM generates intermediate reasoning steps before final answer.

Example:

Step 1
Step 2
Step 3
Answer

9. What is Tree-of-Thought (ToT)?

Answer:

Instead of one reasoning path, explores multiple reasoning paths and selects the best one.

10. What is Reflection in Agents?

Answer:

Agent evaluates its own output, identifies mistakes, and improves the response.

11. What is Agent Memory?

Answer:

Memory stores previous interactions and context.

Types:

Short-term memory
Long-term memory
12. What is Episodic Memory?

Answer:

Stores past experiences/actions.

Example:

User previously preferred Python solutions.

13. What is Semantic Memory?

Answer:

Stores facts and knowledge.

Example:

Paris = capital of France

14. What is Multi-Agent System?

Answer:

Multiple agents collaborate to solve a problem.

Example:

Research Agent
Coding Agent
Testing Agent

15. Benefits of Multi-Agent Systems?

Answer:

Parallel execution
Better specialization
Reduced complexity
Scalable workflows
16. What is a Supervisor Agent?

Answer:

An agent that coordinates multiple sub-agents and combines results.

17. What is Human-in-the-Loop?

Answer:

Human approval is required before critical actions.

Example:

Agent drafts email
Human approves
Email sent

18. What is Agent Orchestration?

Answer:

Managing interactions among multiple agents, tools, and workflows.

19. What is LangGraph?

Answer:

LangGraph is a framework for building stateful, cyclical agent workflows using graphs.

Nodes = Tasks
 Edges = Transitions

20. Why LangGraph over simple chains?

Answer:

Supports:

Loops
Branching
State persistence
Multi-agent workflows
21. What are Nodes in LangGraph?

Answer:

Nodes represent processing steps.

Example:

Planner
Retriever
Generator

22. What are Edges in LangGraph?

Answer:

Edges determine flow from one node to another.

23. What is State in LangGraph?

Answer:

Shared data passed across nodes.

Example:

state = {
 "question": "...",
 "retrieved_docs": "...",
 "answer": "..."
}

24. What is Conditional Routing?

Answer:

Workflow path changes dynamically based on agent output.

Example:

if confidence < 80%
 → retrieve more info
else
 → answer

25. What is MCP?

Answer:

MCP = Model Context Protocol

An open standard that lets LLMs communicate with external tools, resources, and data sources.

26. What problem does MCP solve?

Answer:

Without MCP:

Tool A → custom integration
Tool B → custom integration
Tool C → custom integration


With MCP:

One standard protocol
Many tools

27. MCP Architecture?

Answer:

Client
 ↓
MCP Server
 ↓
Tools
Resources
Prompts

28. What is an MCP Client?

Answer:

Application that connects to MCP servers.

Examples:

Claude Desktop
VS Code
Agent Frameworks
29. What is an MCP Server?

Answer:

Exposes tools, resources, and prompts through MCP protocol.

30. What are MCP Tools?

Answer:

Executable functions.

Examples:

run_sql()
search_docs()
send_email()

31. What are MCP Resources?

Answer:

Readable data sources.

Examples:

PDFs
Files
Databases
Knowledge Bases

32. What are MCP Prompts?

Answer:

Reusable prompt templates exposed by MCP servers.

33. MCP vs API?

Answer:

MCP	APIStandard for AI tools	Generic software interface
Tool discovery	No discovery
LLM-friendly	Developer-oriented
Context-aware	Request-response

Interview Answer: "MCP is to AI tools what USB is to hardware devices."

34. MCP vs RAG?

Answer:

RAG

Retrieves documents.
Provides knowledge.

MCP

Connects tools/resources.
Enables actions.

Example:

RAG → Read database docs
MCP → Execute database query

35. Explain End-to-End Agent Execution Flow.

Answer:

User Query
     ↓
Planner Agent
     ↓
Tool Selection
     ↓
Tool Execution
     ↓
Observation
     ↓
Reasoning
     ↓
If Needed → More Tools
     ↓
Final Response


Example:

User:
What were my sales last month?

Agent:
1. Understand question
2. Choose SQL tool
3. Query database
4. Analyze result
5. Generate answer

Most Important Interview Question

Q: What is the difference between RAG, Agent, and MCP?

Answer:

Technology	PurposeRAG	Retrieve knowledge
Agent	Decide and act
MCP	Standardized connection to tools

Easy Memory Trick

RAG = READ
Agent = THINK + ACT
MCP = CONNECT


This single comparison is asked very frequently in GenAI/Agentic AI assessments.
```
