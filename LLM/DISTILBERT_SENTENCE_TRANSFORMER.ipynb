````python
**DistilBERT uses the *WordPiece* tokenizer algorithm.**  

---

## 🔍 Details
- **Tokenizer type:** DistilBERT inherits its tokenizer from BERT, which is based on **WordPiece subword segmentation**.  
- **How it works:**  
  - Breaks text into subword units rather than whole words.  
  - Common words remain intact (e.g., *machine* → "machine").  
  - Rare or complex words are split into smaller pieces (e.g., *transformers* → "trans", "##form", "##ers").  
  - The `##` prefix indicates that the token is a continuation of a word.  
- **Why WordPiece:**  
  - Keeps vocabulary size manageable (~30k tokens).  
  - Handles out-of-vocabulary words gracefully by splitting them into known subwords.  
  - Improves efficiency and generalization compared to word-level tokenization.  

---

## ⚡ Example Demo
Here’s how DistilBERT tokenizes a sentence:

```python
from transformers import DistilBertTokenizer

tokenizer = DistilBertTokenizer.from_pretrained("distilbert-base-uncased")

text = "AI/ML is transforming careers."
tokens = tokenizer.tokenize(text)

print(tokens)
```

**Output:**
```
['ai', '/', 'ml', 'is', 'transform', '##ing', 'careers', '.']
```

👉 Notice how *transforming* is split into `"transform"` + `"##ing"` — classic WordPiece behavior.

---

## 🧠 Recruiter‑Friendly Note
When explaining DistilBERT in interviews or portfolio projects:
- Emphasize that **DistilBERT is a distilled version of BERT**, so it inherits **WordPiece tokenization**.  
- Highlight how **subword tokenization** solves the out‑of‑vocabulary problem and makes models robust across domains.  
- You can even demo tokenization side‑by‑side with other algorithms (e.g., BPE, SentencePiece) to show your depth of understanding.  

---

Would you like me to create a **side‑by‑side numeric demo** comparing **WordPiece vs BPE vs SentencePiece** on the same sentence? That would be a recruiter‑ready showcase of your NLP mastery.
````
