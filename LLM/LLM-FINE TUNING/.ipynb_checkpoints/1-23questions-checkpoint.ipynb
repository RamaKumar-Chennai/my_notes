```python
Great choice. If the assessment is specifically on LLM Fine-Tuning, most companies don't expect you to be a researcher. They usually test whether you understand:

Why fine-tuning is needed
Different fine-tuning methods
Dataset preparation
LoRA/QLoRA
Training process
Hyperparameters
Evaluation
Fine-tuning vs RAG
Common problems and solutions
Deployment considerations

Since you already worked with Qwen + LoRA and have exposure to RAG, I'll focus on assessment-style preparation.

1. What is Fine-Tuning?

Fine-tuning means taking a pre-trained LLM and training it further on your custom data.

Example

Base model:

What is the capital of India?

Answer:

New Delhi

Now suppose you want a Cricket Commentary Assistant.

You fine-tune the model on:

Input:
Who hit the winning six?

Output:
MS Dhoni hit the winning six in the final over.


After training, the model becomes specialized in cricket-related tasks.

2. Why Not Train From Scratch?

Training from scratch:

Needs:
- Trillions of tokens
- Hundreds of GPUs
- Millions of dollars


Fine-tuning:

Already know language
+
Teach specific domain


Example:

Base Model + Medical Data
= Medical Assistant

Base Model + SQL Data
= Text-to-SQL Model

Base Model + Legal Data
= Legal Assistant

3. Types of Fine-Tuning
Full Fine-Tuning

Train all model weights.

7B model
= billions of parameters updated


Pros:

Highest performance


Cons:

Very costly

PEFT

Parameter Efficient Fine-Tuning

Only train a small subset.

Examples:

LoRA
QLoRA
Adapters
Prefix Tuning
Prompt Tuning


Most assessments focus on:

LoRA
QLoRA

4. LoRA (VERY IMPORTANT)

Almost guaranteed assessment topic.

Problem

Suppose model has:

7 Billion parameters


Training all parameters is expensive.

Solution

Freeze original weights.

Train only small matrices.

Concept:

New Weight
=
Original Weight + Low Rank Update


Instead of changing huge matrix W:

W


LoRA learns:

A × B


where:

Rank r << Original Dimension


Example:

4096 x 4096 matrix

instead of training:
16 million values

LoRA trains:
4096x8 + 8x4096


Much smaller.

5. LoRA Parameters
Rank (r)

Controls adaptation capacity.

r=4
small learning

r=16
moderate

r=64
large adaptation


Higher rank:

More learning
More memory

Alpha

LoRA scaling factor.

Formula:

scale = alpha/r


Common:

r=8
alpha=16

Target Modules

Where LoRA is applied.

Usually:

target_modules=[
"q_proj",
"k_proj",
"v_proj",
"o_proj"
]


Assessment Question:

Why apply LoRA to q_proj and v_proj?

Answer:

These attention layers strongly influence model behavior while requiring fewer trainable parameters.

6. QLoRA

Most popular today.

Question appears very often.

LoRA
Model in 16-bit

QLoRA
Model in 4-bit
+
LoRA adapters


Benefits:

Much less GPU memory


Example:

7B model


Before:

14-16GB+


After QLoRA:

4-6GB


Key Idea:

Quantize model
Train LoRA adapters

7. Quantization

Reducing precision.

FP32
32 bits

FP16
16 bits

INT8
8 bits

INT4
4 bits


Less precision:

Less memory
Faster inference


Trade-off:

Slight accuracy loss

8. Instruction Fine-Tuning

Most commonly used.

Data format:

{
 "instruction":"Translate English to French",
 "input":"Hello",
 "output":"Bonjour"
}


Training teaches:

Follow human instructions

9. Supervised Fine-Tuning (SFT)

Also important.

Dataset:

Prompt -> Ideal Response


Example:

Question:
What is AI?

Answer:
Artificial Intelligence...


Model learns:

Input → Output


Most open-source fine-tuning uses SFT.

10. RLHF

Reinforcement Learning from Human Feedback

Pipeline:

Pretraining
      ↓
SFT
      ↓
Reward Model
      ↓
RLHF


Purpose:

Make responses align with humans


ChatGPT uses RLHF-style alignment.

11. Dataset Preparation

Critical topic.

Dataset quality > dataset size.

Good dataset:

Correct
Consistent
Representative
Clean


Bad dataset:

Duplicates
Typos
Conflicting responses

12. Train / Validation / Test

Typical split:

Train      80%
Validation 10%
Test       10%


Purpose:

Train

Learn.

Validation

Tune hyperparameters.

Test

Final evaluation.

13. Hyperparameters
Learning Rate

Step size of learning.

Too High:

Unstable


Too Low:

Slow learning


Common:

1e-5
2e-5
5e-5

Batch Size

Examples processed together.

Higher batch size
=
More memory

Epoch

One complete pass through dataset.

1000 records

1 epoch
=
See all 1000 once

14. Overfitting

Model memorizes training data.

Symptoms:

Train loss ↓
Validation loss ↑


Assessment favorite.

Solution:

More data
Regularization
Early stopping
Fewer epochs

15. Underfitting

Model not learning enough.

Symptoms:

Train loss high
Validation loss high


Solutions:

More epochs
Better data
Larger model

16. Loss Function

Measures error.

Goal:

Minimize Loss


Training Curve:

Epoch 1  Loss=3.2
Epoch 2  Loss=2.1
Epoch 3  Loss=1.0


Good sign.

17. Fine-Tuning vs RAG

Extremely common interview question.

Fine-Tuning	RAGTeaches model new behavior	Provides external knowledge
Changes weights	Doesn't change weights
Expensive	Cheaper
Static knowledge	Dynamic knowledge
Retraining needed	Simply update documents
Example

Company handbook changes every week.

Use:

RAG


Not:

Fine-tuning

18. When To Use Fine-Tuning?

Use when:

Need specific style
Need task specialization
Need format control
Need domain behavior


Examples:

SQL generation
Medical report generation
Customer support style
Legal drafting

19. When Not To Use Fine-Tuning?

Bad reasons:

Want latest news
Want changing data
Want database lookups


Use:

RAG


instead.

20. Evaluation Metrics
Perplexity

Measures prediction quality.

Lower:

Better

BLEU

Translation tasks.

ROUGE

Summarization.

Accuracy

Classification.

21. Catastrophic Forgetting

Very important.

Model learns new task but forgets old knowledge.

Example:

Fine-tune heavily on law data

Model gets worse at general tasks.


Solution:

Mixed datasets
LoRA
Careful training

22. Hallucination After Fine-Tuning

Fine-tuning does NOT eliminate hallucinations.

Model can still:

Invent facts
Generate incorrect answers


Use:

RAG + Fine-tuning


for best results.

23. Complete Training Pipeline
Collect Data
      ↓
Clean Data
      ↓
Create Train/Val/Test
      ↓
Choose Base Model
      ↓
Apply LoRA/QLoRA
      ↓
Train
      ↓
Evaluate
      ↓
Deploy
      ↓
Monitor

Assessment "Must Remember" Questions
What is LoRA?
Why LoRA is memory efficient?
Difference between LoRA and QLoRA?
What is quantization?
What is instruction tuning?
What is SFT?
What is RLHF?
Train vs Validation vs Test?
What is overfitting?
Fine-tuning vs RAG?
What is catastrophic forgetting?
What is learning rate?
What is rank in LoRA?
Why use q_proj/v_proj?
Why not train from scratch?

Master these topics first. They typically cover 70-80% of an LLM Fine-Tuning assessment. After that, we can move to advanced topics like tokenization, attention mechanism, gradient accumulation, mixed precision, checkpointing, packing, instruction templates, DPO, PPO, reward models, and scenario-based fine-tuning questions, which are where harder assessment questions usually
    come from.
```
