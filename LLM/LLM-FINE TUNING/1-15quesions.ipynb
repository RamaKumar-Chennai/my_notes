```python
These are interview-style answers. Try to explain them in your own words rather than memorizing.

1. What is LoRA?

LoRA (Low-Rank Adaptation) is a Parameter-Efficient Fine-Tuning (PEFT) technique where we freeze the original model weights and train only small 
adapter matrices.

Instead of changing:

WW

LoRA learns:

W+BAW + BA

where A and B are small trainable matrices.

Interview Answer

LoRA is a fine-tuning method that freezes the base model and trains small low-rank adapter matrices, significantly reducing memory and compute 
requirements.

2. Why is LoRA memory efficient?

In full fine-tuning, all model parameters are updated.

Example:

7B model
→ 7 billion trainable parameters


In LoRA:

7B model
→ Base weights frozen
→ Only adapter weights trained


Results:

Less GPU memory
Faster training
Smaller checkpoints
Interview Answer

LoRA is memory efficient because only a small number of adapter parameters are trained while the original model weights remain frozen.

3. Difference Between LoRA and QLoRA?
LoRA
Base model stored in FP16/BF16
Adapter weights trained
QLoRA
Base model quantized to 4-bit
Adapter weights trained
Memory Usage
Full Fine-Tuning
      ↓
LoRA
      ↓
QLoRA

Interview Answer

QLoRA combines LoRA with 4-bit quantization of the base model, reducing memory usage even further while maintaining performance.

4. What is Quantization?

Quantization reduces the precision of model weights.

Example:

FP32 → 32-bit
FP16 → 16-bit
INT8 → 8-bit
4-bit → 4-bit


Benefits:

Less memory
Faster inference
Lower hardware requirements
Interview Answer

Quantization is the process of storing model weights using fewer bits to reduce memory and computational requirements.

5. What is Instruction Tuning?

Instruction tuning teaches a model how to follow instructions.

Dataset Example:

{
 "instruction":"Translate to French",
 "response":"Bonjour"
}


The model learns:

Instruction → Correct Response

Interview Answer

Instruction tuning is the process of fine-tuning a model on instruction-response pairs so it learns to follow human instructions more effectively.

6. What is SFT?

SFT = Supervised Fine-Tuning

Model learns from labeled examples.

Example:

{
 "question":"Capital of France?",
 "answer":"Paris"
}


Loss is calculated against the known answer.

Interview Answer

SFT is a supervised training stage where the model learns from high-quality input-output examples.

7. What is RLHF?

RLHF = Reinforcement Learning from Human Feedback

Steps:

Step 1

Pretrained Model

↓

Step 2

Supervised Fine-Tuning

↓

Step 3

Humans rank responses

Example:

Response A ✅
Response B ❌


↓

Step 4

Reward Model learns preferences

↓

Step 5

Reinforcement Learning optimizes the LLM

Interview Answer

RLHF uses human feedback to train a reward model, which is then used to improve model responses through reinforcement learning.

8. Train vs Validation vs Test
Training Set

Used to learn patterns.

Validation Set

Used during training to monitor performance and tune hyperparameters.

Test Set

Used only for final evaluation.

Example:

1000 samples

Train = 700
Validation = 150
Test = 150

Interview Answer

The training set teaches the model, the validation set helps tune it, and the test set provides an unbiased final evaluation.

9. What is Overfitting?

Overfitting happens when the model memorizes training data instead of learning general patterns.

Example:

Train Accuracy = 99%

Test Accuracy = 70%


Signs:

Training loss ↓
Validation loss ↑
Interview Answer

Overfitting occurs when a model performs very well on training data but poorly on unseen data because it has memorized rather than generalized.

10. Fine-Tuning vs RAG
Fine-Tuning	RAGChanges model behavior	Retrieves external knowledge
Updates weights	Doesn't change weights
Good for style/format	Good for dynamic data
Requires training	No training needed
Example

Need company policies updated daily?

✅ RAG

Need answers always in JSON?

✅ Fine-Tuning

Interview Answer

Fine-tuning teaches a model new behavior, while RAG provides external knowledge at inference time without modifying model weights.

11. What is Catastrophic Forgetting?

During fine-tuning, the model may forget knowledge learned during pretraining.

Example:

Model originally:

Excellent at coding


After aggressive medical fine-tuning:

Medical questions good
Coding performance drops

Interview Answer

Catastrophic forgetting occurs when a fine-tuned model loses previously learned capabilities while adapting to new tasks.

12. What is Learning Rate?

Learning rate controls how much weights change during each update.

Example:

learning_rate = 2e-5


Think of it as step size.

Too High
Training unstable

Too Low
Training very slow

Interview Answer

Learning rate determines the size of weight updates during training and directly affects convergence speed and stability.

13. What is Rank in LoRA?

Rank (r) determines the size of LoRA adapters.

Example:

r = 4
r = 8
r = 16
r = 32

Small Rank
Less memory
Less learning capacity
Large Rank
More memory
More learning capacity
Interview Answer

Rank controls the dimensionality of LoRA adapter matrices and determines the trade-off between model capacity and memory usage.

14. Why use q_proj and v_proj?

Transformer attention contains:

q_proj
k_proj
v_proj
o_proj


Most information flow happens through attention.

Updating:

q_proj
v_proj


often gives good performance with minimal parameters.

q_proj

Creates query vectors.

v_proj

Creates value vectors.

Interview Answer

q_proj and v_proj are commonly targeted because adapting attention mechanisms provides strong performance gains while keeping the number of 
trainable parameters small.

15. Why not train from scratch?

Training from scratch requires:

Massive datasets
Huge compute resources
Long training times
Millions of dollars for large models

Example:

Pretraining Llama/GPT
→ Trillions of tokens
→ Thousands of GPUs


Fine-tuning uses existing knowledge.

Interview Answer

We do not train from scratch because pretrained models already contain language knowledge, so fine-tuning is much faster, cheaper, and requires 
far less data.

Very Common Follow-Up Question
"Explain the complete pipeline from Pretraining to RLHF."
Large Text Corpus
       ↓
Pretraining
       ↓
Base Model
       ↓
SFT (Instruction Tuning)
       ↓
Instruction Following Model
       ↓
Human Preference Data
       ↓
Reward Model
       ↓
RLHF
       ↓
ChatGPT-like Assistant


This pipeline is one of the most frequently asked Fine-Tuning interview questions.
```
