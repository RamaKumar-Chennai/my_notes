```python
Perfect for interview preparation. Don't memorize one-line definitions. Understand the concept because interviewers usually ask follow-ups.

1. What is fine-tuning?

Fine-tuning is the process of taking a pretrained model and training it further on a task-specific dataset.

Example:

Base Model: Llama, Qwen
Fine-tuned Model: Customer Support Bot

The model's weights are updated using new training data.

2. Why fine-tune a model instead of using prompt engineering?

Prompt engineering only guides the model temporarily.

Fine-tuning permanently changes model behavior.

Example:

Prompt:

Always answer in JSON format.


Model may still fail sometimes.

Fine-tuning can make the model consistently generate JSON.

3. What is the difference between pretraining and fine-tuning?
Pretraining
Learns general language patterns.
Massive datasets.
Very expensive.

Example:

Wikipedia
Books
Web Pages

Fine-Tuning
Learns a specific task.
Small dataset.
Much cheaper.

Example:

Customer Support QA
SQL Generation
Medical Classification

4. When should you choose RAG instead of fine-tuning?

Use RAG when knowledge changes frequently.

Examples:

Company documents
Policies
Product catalogs
SQL schemas

Fine-tuning cannot update knowledge quickly.

5. When should you use both RAG and fine-tuning together?

Use Fine-Tuning for behavior and RAG for knowledge.

Example:

Fine-Tuning:
Professional tone

RAG:
Latest company documents


This is common in enterprise AI systems.

6. What types of tasks benefit most from fine-tuning?
Classification
Summarization
Translation
SQL Generation
Chatbots
Sentiment Analysis
Domain-specific assistants
7. Can fine-tuning teach a model new knowledge?

Partially.

Small knowledge additions may work.

Large dynamic knowledge should use RAG.

Interview Answer:

Fine-tuning is best for learning behavior, formatting, and task patterns rather than storing large amounts of changing knowledge.

8. Does fine-tuning modify model weights?

Yes.

The model updates weights during training.

LoRA updates only small adapter weights.

Full Fine-Tuning updates all weights.

9. Why is fine-tuning expensive for large models?

Because:

Billions of parameters
Large GPU memory
Long training time
Higher compute cost

Example:

7B model
13B model
70B model


Training cost grows significantly.

10. What happens if training data is poor quality?

Results:

Hallucinations
Wrong outputs
Overfitting
Poor generalization

Garbage In → Garbage Out.

11. What is an instruction-response dataset?

Dataset format:

{
 "instruction":"Translate to French",
 "response":"Bonjour"
}


The model learns how to respond to instructions.

12. Why should training data be diverse?

Prevents memorization.

Helps model generalize to unseen questions.

More diversity = better robustness.

13. What is data leakage?

Information from validation/test data accidentally appears in training data.

Result:

Artificially high performance.

Very bad evaluation practice.

14. Why split data into train, validation, and test sets?
Train

Learn patterns.

Validation

Tune hyperparameters.

Test

Final unbiased evaluation.

15. What happens if validation loss increases while training loss decreases?

The model is overfitting.

It memorizes training examples but performs poorly on unseen data.

16. What is overfitting?

Model memorizes training data.

Example:

Train Accuracy = 99%
Test Accuracy = 70%


Poor generalization.

17. What is underfitting?

Model has not learned enough.

Example:

Train Accuracy = 55%
Test Accuracy = 52%


Bad performance everywhere.

18. How do you clean fine-tuning datasets?
Remove duplicates
Remove incorrect labels
Fix formatting
Remove noisy records
Standardize responses
19. What is data augmentation?

Creating additional training samples.

Example:

Original:

I love this product.


Augmented:

This product is amazing.


Increases dataset diversity.

20. Why remove duplicate examples?

Duplicates cause memorization and bias.

Model may overfit.

21. What is LoRA?

Low-Rank Adaptation.

Instead of updating all weights:

W → W + ΔW


LoRA trains a small low-rank matrix.

Much cheaper.

22. Why is LoRA more efficient than full fine-tuning?

Because:

Fewer trainable parameters
Lower GPU memory
Faster training
Smaller checkpoints
23. What is PEFT?

Parameter Efficient Fine-Tuning.

Methods:

LoRA
QLoRA
Prompt Tuning
Prefix Tuning

Goal:

Update as few parameters as possible.

24. What does rank (r) mean in LoRA?

Rank defines adapter size.

Example:

r=4
r=8
r=16
r=32


Higher rank:

More learning capacity
More memory usage
25. What is lora_alpha?

Scaling factor.

Formula:

Output = LoRA × alpha/r


Controls LoRA influence.

26. What is lora_dropout?

Dropout applied to LoRA layers during training.

Purpose:

Reduce overfitting.

Common:

0.05
0.1

27. Which layers are commonly targeted in LoRA?

Transformer attention layers:

q_proj
k_proj
v_proj
o_proj


Often:

q_proj
v_proj

28. Why freeze most model weights in LoRA?

To:

Save memory
Reduce compute
Preserve pretrained knowledge

Only adapters learn.

29. How much memory can LoRA save compared to full fine-tuning?

Typically 10x-100x less trainable parameters.

Interview answer:

LoRA dramatically reduces memory usage because only adapter weights are trained instead of all model weights.

30. What happens if rank is too small or too large?
Too Small

Cannot learn enough.

Underfitting.

Too Large

Higher memory.

Possible overfitting.

31. What is a checkpoint?

Saved model during training.

Contains:

Weights
Optimizer state
Training progress

Allows resume/recovery.

32. What is an epoch?

One complete pass through the dataset.

Example:

1000 samples


Seen once = 1 epoch.

33. What is a batch size?

Number of samples processed before weight update.

Example:

Batch Size = 8


Process 8 examples together.

34. What is gradient accumulation?

Accumulates gradients across multiple batches before updating weights.

Simulates larger batch sizes.

35. Why use gradient accumulation?

When GPU memory is limited.

Example:

Batch Size = 4
Accumulation = 8

Effective Batch = 32

36. What is a learning rate?

Controls step size during weight updates.

Most important hyperparameter.

Example:

learning_rate = 2e-5

37. What happens if learning rate is too high?
Training unstable
Loss oscillates
Model may diverge
38. What happens if learning rate is too low?
Very slow training
May not converge
Wastes compute
39. What is warmup?

Gradually increases learning rate at the beginning of training.

Benefits:

Stable training
Prevents large early updates
40. Why use weight decay?

Regularization technique.

Prevents excessively large weights.

Reduces overfitting.

41. What is training loss?

Error calculated on training data.

Lower is generally better.

42. What is validation loss?

Error calculated on unseen validation data.

Measures generalization.

43. Why can training loss decrease while model quality worsens?

Because model starts memorizing.

Training loss ↓

Validation loss ↑

This indicates overfitting.

44. What is perplexity?

Measures how well a language model predicts text.

Formula:

Perplexity = exp(loss)


Lower is better.

45. Lower perplexity means what?

Better predictions.

Example:

PPL = 8


is better than

PPL = 20

46. What metrics would you use?
Classification
Accuracy
Precision
Recall
F1-Score
Summarization
ROUGE
Question Answering
Exact Match (EM)
F1
Translation
BLEU
47. What is early stopping?

Stop training when validation performance stops improving.

Prevents overfitting.

48. Why save best checkpoint instead of latest checkpoint?

Latest checkpoint may already be overfitted.

Best checkpoint has lowest validation loss.

49. How do you evaluate a fine-tuned LLM?
Validation loss
Perplexity
Task metrics
Human evaluation
Benchmark datasets

Example:

SQL correctness
Answer quality
Classification accuracy
50. How do you know fine-tuning was successful?

Indicators:

✅ Lower validation loss
 ✅ Better task performance
 ✅ Better benchmark scores
 ✅ Better human evaluation
 ✅ Generalizes to unseen examples

Interview One-Liner

Fine-tuning is successful when the model performs significantly better on the target task while maintaining good generalization on unseen data.

For assessment preparation, Questions 4, 5, 15, 21, 24, 25, 34, 39, 43, 44, 47, and 49 are the most frequently asked follow-up questions. These are the ones interviewers usually explore in depth.
```
