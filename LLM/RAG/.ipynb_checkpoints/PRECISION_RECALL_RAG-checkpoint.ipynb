```python
Let's focus only on Scenario 2.

Suppose there are actually 10 relevant documents in the entire database:

Relevant docs in DB:

D1 D2 D3 D4 D5 D6 D7 D8 D9 D10


These are the only documents that contain the answer.

Now your retriever returns:

D1 D2 D3 D4 D5 X1 X2 X3 X4 X5


where:

D1-D5 = Relevant
X1-X5 = Irrelevant


So the retriever returned 10 documents total.

Out of those 10:

5 are relevant
5 are irrelevant

Precision

Precision asks:

Of the documents I retrieved, how many are actually relevant?

Retrieved:

10 docs


Relevant among retrieved:

5 docs


Therefore:

Precision = 5/10 = 50%


Makes sense.

Recall

Recall asks a completely different question:

Out of ALL relevant documents available in the database, how many did I find?

Let's count.

Relevant docs existing in DB:

D1 D2 D3 D4 D5 D6 D7 D8 D9 D10


Total relevant docs:

10


How many of those did the retriever find?

Retrieved:

D1 D2 D3 D4 D5


Found:

5 relevant docs


Missed:

D6 D7 D8 D9 D10


Therefore:

Recall = 5/10 = 50%

Why don't X1-X5 affect Recall?

This is usually the confusing part.

Recall only cares about:

Did you find the relevant documents?


It does NOT care how much junk came along.

Suppose retriever returned:

D1 D2 D3 D4 D5
+
1000 irrelevant docs


Then:

Precision = 5/1005


Very poor.

But Recall is still:

5/10


because you still found only 5 of the 10 relevant documents.

Easy Cricket Example 😄

Suppose there are 10 sixes hit in a match.

Your retrieval system is trying to find all sixes.

Actual sixes:

Ball1
Ball2
Ball3
Ball4
Ball5
Ball6
Ball7
Ball8
Ball9
Ball10


Your system returns:

Ball1
Ball2
Ball3
Ball4
Ball5
+
5 normal dot balls

Recall

Question:

Out of the 10 actual sixes, how many did you find?

Answer:

5


Therefore:

Recall = 5/10 = 50%

Precision

Question:

Out of the 10 balls returned, how many are really sixes?

Answer:

5


Therefore:

Precision = 5/10 = 50%

Memory Trick

Recall = "What did I MISS?"

Recall =
Relevant docs found
-------------------
All relevant docs


Precision = "How much JUNK did I bring?"

Precision =
Relevant docs found
-------------------
All retrieved docs


So:

Recall → cares about missed relevant docs
Precision → cares about retrieved irrelevant docs ✅
```

```python
This is one of the most confusing concepts in RAG. Let's make it super simple.

Suppose for a question there are actually 10 relevant documents in your database.

Relevant Docs:
D1 D2 D3 D4 D5 D6 D7 D8 D9 D10


These are the documents that truly contain useful information.

Scenario 1

Retriever returns:

D1 D2 D3 D4 D5


All 5 are relevant.

Precision

Formula:

Relevant Retrieved Docs
-----------------------
Retrieved Docs


Here:

Relevant Retrieved Docs = 5
Retrieved Docs = 5


So:

Precision = 5/5 = 100%


Meaning:

Everything retrieved was useful.

Recall

Formula:

Retrieved Relevant Docs
-----------------------
All Relevant Docs


Here:

Retrieved Relevant Docs = 5
All Relevant Docs = 10


So:

Recall = 5/10 = 50%


Meaning:

We found only 5 of the 10 useful documents.

Scenario 2

Retriever returns:

D1 D2 D3 D4 D5 X1 X2 X3 X4 X5


where:

D1-D5 = relevant
X1-X5 = irrelevant

Precision
Relevant Retrieved Docs = 5
Retrieved Docs = 10

Precision = 5/10 = 50%


Meaning:

Half the retrieved documents are junk.

Recall
Retrieved Relevant Docs = 5
All Relevant Docs = 10

Recall = 5/10 = 50%


Recall didn't change.

Why?

Because you still found only 5 of the 10 useful documents.

Scenario 3

Retriever returns:

D1 D2 D3 D4 D5 D6 D7 D8 D9 D10

Precision
10/10 = 100%

Recall
10/10 = 100%


Perfect retrieval.

The Key Difference
Precision asks:

Out of what I retrieved, how much was correct?

Precision =
Relevant Retrieved Docs
-----------------------
Total Retrieved Docs


Focus:

Quality of retrieved results

Recall asks:

Out of all useful documents available, how many did I find?

Recall =
Retrieved Relevant Docs
-----------------------
All Relevant Docs


Focus:

Coverage

Easy Interview Example

Suppose there are 100 relevant documents in the database.

Your retriever returns:

20 documents


Among those 20:

15 are relevant
5 are irrelevant

Precision
15/20
= 75%


Meaning:

75% of retrieved documents are useful.

Recall
15/100
= 15%


Meaning:

Out of all 100 useful documents, only 15 were found.

Memory Trick
Precision
Retrieved → Check Quality

Relevant Retrieved
------------------
Retrieved


"Of what I got, how much is useful?"

Recall
Relevant → Check Coverage

Relevant Retrieved
------------------
All Relevant


"Of all useful docs, how many did I find?"

For RAG interviews, a common one-line answer is:

Precision measures the accuracy of retrieved documents, while Recall measures how completely the retriever finds all relevant documents.
                 High Precision means less junk; High Recall means fewer missed relevant documents. ✅
```
