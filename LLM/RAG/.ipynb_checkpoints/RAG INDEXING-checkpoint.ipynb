```python
Great question. Let's understand it with a simple example.

Without Indexing (Brute Force Search)

Suppose your vector database contains 1 million document vectors.

User asks:

What is LangGraph?


The query is converted into a vector:

Q = [0.2, 0.8, 0.4, ...]


Now FAISS/vector DB must find the most similar vectors.

What happens?

It compares Q with:

Doc1
Doc2
Doc3
...
Doc1000000


One by one.

Q vs Doc1
Q vs Doc2
Q vs Doc3
...
Q vs Doc1000000


Calculates cosine similarity for every single vector.

This is called:

Exact Nearest Neighbor Search

Problem

For 1 million vectors:

1,000,000 similarity calculations


For 100 million vectors:

100,000,000 calculations


Very slow.

With Indexing

Before searching, vectors are organized intelligently.

Think of a library.

Without Index

You enter a library and ask:

Find a Python book.


Librarian checks:

Shelf 1
Shelf 2
Shelf 3
...
Shelf 10000


Very slow.

With Index

Books are categorized:

Programming
History
Science
Sports


Librarian immediately goes to:

Programming


Then searches only there.

Much faster.

IVF (Inverted File Index)

Suppose we have:

1 million vectors


FAISS groups them into clusters.

Cluster A
Cluster B
Cluster C
Cluster D
...


Example:

A -> Sports
B -> Movies
C -> AI
D -> Finance


Query:

LangGraph


Embedding is closest to:

Cluster C (AI)


Search only inside Cluster C.

Instead of:

1,000,000 vectors


Maybe search:

10,000 vectors


Huge speed improvement.

HNSW (Hierarchical Navigable Small Worlds)

Think of Google Maps.

Without HNSW:

Try every possible road.


With HNSW:

Follow important highways first,
then local roads.


Vectors are connected like a graph.

DocA ---- DocB
  |         |
  |         |
DocC ---- DocD


Query enters graph and "jumps" toward similar vectors.

Instead of checking all vectors:

1,000,000


It may check only:

200


or

500


vectors.

Very fast.

PQ (Product Quantization)

Problem:

Each vector can be large.

Example:

1536 dimensions


Storing millions of these vectors requires huge memory.

PQ compresses vectors.

Instead of storing:

[0.2345, 0.7854, 0.4567, ...]


Store compact codes:

A12 B45 C22


Benefits:

✅ Less memory

✅ Faster search

❌ Slightly less accuracy

Interview Answer

What is Vector Indexing?

Vector indexing is the process of organizing embeddings so that similarity search can be performed efficiently. Without indexing, the system compares 
the query vector with every stored vector, which becomes very slow for large datasets. Index structures such as IVF, HNSW, and PQ reduce the search 
space and enable fast approximate nearest-neighbor retrieval while maintaining high accuracy.

One-Line Memory Trick
Without Indexing:
Search All Vectors

With Indexing:
Search Only Smartly Chosen Vectors


For GenAI interviews, remember:

No index → Exact search → Slow
IVF → Cluster-based search
HNSW → Graph-based navigation
PQ → Compressed vectors for faster search and less memory
```

```python
Yes. The centroid is calculated during the index-building phase, and which system does it depends on the vector index implementation.

For FAISS IVF indexes, FAISS itself computes the centroids using a clustering algorithm (typically k-means) when you train the index.

Example:

index = faiss.IndexIVFFlat(
    quantizer,
    dimension,
    nlist=100
)

index.train(vectors)


During:

index.train(vectors)


FAISS does:

Takes all vectors.
Runs k-means clustering.
Creates 100 clusters (nlist=100).
Computes 100 centroid vectors.
Stores those centroids in the index.

After training:

Cluster 1 -> Centroid C1
Cluster 2 -> Centroid C2
...
Cluster 100 -> Centroid C100


Then when you do:

index.add(vectors)


each vector is assigned to its nearest centroid.

Is this only FAISS?

No.

Any vector database that uses an IVF-style index must create centroids somehow.

Examples:

FAISS IVF → FAISS computes centroids.
Milvus IVF → Milvus computes centroids.
Pinecone IVF-based implementations → Pinecone computes centroids internally.
Weaviate/Milvus HNSW → No centroids because HNSW uses a graph structure instead.

So centroids are not a property of vector databases in general. They are specifically a property of cluster-based indexes like IVF.

Visual Example

Suppose vectors are:

(1,1)
(1,2)
(2,1)


FAISS may create a cluster:

Cluster A:
(1,1)
(1,2)
(2,1)


Centroid is approximately the average:

C = (
 (1+1+2)/3,
 (1+2+1)/3
)

C = (1.33, 1.33)


This centroid is what FAISS compares against first.

Query
  ↓
Compare with centroids
  ↓
Choose nearest clusters
  ↓
Search vectors inside those clusters


This is why IVF is much faster than scanning every vector.

A common interview question is:

"Are centroid vectors actual document vectors?"

Answer: No.

Centroids are usually artificial vectors calculated as cluster representatives (cluster centers). They often do not correspond to any real document 
embedding.
```
