```python
Excellent preparation strategy, Rama. In many GenAI assessments, the advanced descriptive questions (21-50) are where candidates lose marks. Below are interview-level answers.

21. Difference between Fine-Tuning and RAG?
Fine-Tuning	RAG
Updates model weights	Keeps model unchanged
Requires training	No training required
Expensive	Cheaper
Knowledge becomes part of model	Knowledge stays in documents
Hard to update	Easy to update

Example:

New company policies every week → RAG
Teaching model a writing style → Fine-tuning
22. Why is Hybrid Search often superior?

Hybrid Search combines:

Semantic Search (Vector)
+
Keyword Search (BM25)


Example:

Query:

GPT-4o pricing


Vector search understands meaning.

BM25 ensures exact terms like:

GPT-4o
pricing


are matched.

Result: Better retrieval accuracy.

23. What is Contextual Compression?

Many retrieved chunks contain irrelevant text.

Example:

Retrieved chunk:

Company founded in 2000.
Revenue was 10M.
CEO is John.


Question:

Who is CEO?


Compressed chunk:

CEO is John.


Reduces token usage and improves answer quality.

24. Explain Parent-Child Retrieval

Documents are split into:

Parent Chunk
     |
Child Chunks


Example:

Parent:

Entire chapter


Children:

paragraph1
paragraph2
paragraph3


Search happens on child chunks.

Returned context is parent chunk.

Benefits:

Accurate retrieval
Richer context
25. What is GraphRAG?

GraphRAG stores entities and relationships.

Example:

Rama
  |
works_on
  |
Cricket Project


Graph:

Rama --> Cricket Project
Project --> SQL Agent
Project --> RAG Agent


Advantage:

Can answer multi-hop reasoning queries.

Example:

Which project uses SQL agent?

26. Explain Multi-Query Retrieval

Instead of one query:

What is LangGraph?


Generate multiple versions:

LangGraph architecture
LangGraph workflow
LangGraph agents
LangGraph state management


Retrieve for all versions.

Merge results.

Advantage:

Improves Recall.

27. What does Reranking solve?

Retriever may return:

Doc A -> 0.81
Doc B -> 0.80
Doc C -> 0.79


Sometimes ranking is poor.

Reranker re-evaluates relevance using a stronger model.

Before:

A B C


After:

B A C


Improves precision.

28. What is Embedding Dimensionality?

Length of embedding vector.

Example:

[0.2, 0.5, 0.8]


Dimension = 3

Real models:

384
768
1024
1536
3072


Higher dimensions:

✅ More information

❌ More storage

❌ Slower search

29. Why can larger chunks hurt retrieval?

Large chunk:

1000 words


Contains many topics.

Query may match only one sentence.

Embedding becomes diluted.

Result:

Lower relevance


Smaller chunks improve retrieval precision.

30. What is Vector Indexing?

Technique for fast similarity search.

Without indexing:

Compare query against
every vector.


Too slow.

Index structures:

IVF
HNSW
PQ


Allow approximate nearest-neighbor search.

31. Role of Metadata Filters?

Restrict retrieval.

Example:

{
 "year":2025,
 "department":"HR"
}


Query:

Leave policy


Search only HR documents from 2025.

Improves precision and speed.

32. Why does RAG reduce hallucinations?

Without RAG:

Question -> LLM Memory


With RAG:

Question
   ↓
Retrieve Documents
   ↓
Context
   ↓
LLM


Model answers from actual documents instead of memory.

33. What is Retrieval Recall?

Measures how many relevant documents were found.

Formula:

Recall =
Retrieved Relevant Docs
----------------------
All Relevant Docs


High recall:

Found most useful documents.

34. What is Retrieval Precision?

Measures relevance of retrieved documents.

Formula:

Precision =
Relevant Retrieved Docs
-----------------------
Retrieved Docs


High precision:

Most documents are useful.

35. What is Grounded Generation?

Generated answer is supported by retrieved evidence.

Example:

Document:

CEO = John


Answer:

CEO is John.


Grounded.

If answer says:

CEO is Smith


Not grounded.

36. Difference between Dense and Sparse Retrieval?
Dense Retrieval

Uses embeddings.

Meaning-based search


Example:

car ≈ automobile

Sparse Retrieval

Uses keywords.

Example:

BM25
TF-IDF


Looks for exact words.

37. When should RAG not be used?

Do NOT use RAG when:

Knowledge rarely changes
Information is already in trained model
Fast response is critical
Dataset is tiny

Example:

Basic math


No need for RAG.

38. What is Agentic RAG?

Agents decide:

What to retrieve
Which tools to call
Whether more retrieval is needed

Example:

User Query
   ↓
Agent
   ↓
Search
   ↓
Evaluate
   ↓
Search Again
   ↓
Answer


More intelligent than traditional RAG.

39. What is Query Rewriting?

Reformulating query before retrieval.

User:

How much did they earn?


Rewritten:

How much revenue did Microsoft earn?


Improves retrieval quality.

40. Why use BM25 with vectors?

Vectors capture meaning.

BM25 captures exact keywords.

Combining both:

✅ Better precision

✅ Better recall

✅ Better handling of rare terms

41. Explain Cosine Similarity Mathematically

Formula:

A · B
------------------
||A|| × ||B||


Measures angle between vectors.

Values:

1   -> Identical
0   -> Unrelated
-1  -> Opposite


Most popular similarity metric.

42. What is Maximum Marginal Relevance (MMR)?

Balances:

Relevance
+
Diversity


Without MMR:

5 nearly identical chunks


With MMR:

5 relevant but different chunks


Reduces redundancy.

43. Why is Chunk Overlap used?

Without overlap:

Chunk 1:

Virat scored


Chunk 2:

a century yesterday


Meaning gets split.

With overlap:

Chunk1: Virat scored a century
Chunk2: scored a century yesterday


Preserves context.

Common overlap:

10-20%

44. What is Lost-in-the-Middle Problem?

LLMs often pay less attention to information located in the middle of long contexts.

Example:

Beginning → Important
Middle → Missed
End → Important


Solution:

Compression
Reranking
Better prompt structure
45. What is Semantic Chunking?

Split based on meaning rather than character count.

Example:

Instead of:

500 characters


Split at:

Topic boundaries


Result:

More meaningful retrieval.

46. What is a Reranker Model?

Special model that scores relevance.

Pipeline:

Retriever
   ↓
Top 20 Docs
   ↓
Reranker
   ↓
Top 5 Docs


Examples:

BGE Reranker
Cohere Reranker
Cross Encoder

47. How does LangChain Retriever work?

Example:

retriever = vectorstore.as_retriever()


Process:

Query
 ↓
Embedding
 ↓
Similarity Search
 ↓
Top Documents


Returns relevant documents to chain.

48. What are common RAG Evaluation Metrics?
Retrieval Metrics
Precision
Recall
MRR
Hit Rate
Generation Metrics
Faithfulness
Groundedness
Relevancy
Correctness
49. How would you improve a poorly performing RAG system?
Improve chunking
Add chunk overlap
Use better embeddings
Increase top-k
Add hybrid search
Add reranker
Add metadata filters
Use query rewriting
Improve prompts
Evaluate retrieval metrics

This is a very common interview question.

50. Design an Enterprise RAG Architecture for 1 Million Documents
Documents
     |
Document Loader
     |
Chunking
     |
Embeddings
     |
Vector Database
(FAISS/Pinecone)
     |
Retriever
     |
Hybrid Search
     |
Reranker
     |
Prompt Builder
     |
LLM
     |
Answer + Citations


Advanced Enterprise Version:

Users
   |
API Gateway
   |
Query Rewriter
   |
Hybrid Retriever
(Vector + BM25)
   |
Reranker
   |
Context Compression
   |
LLM
   |
Guardrails
   |
Monitoring
   |
Answer

Interview Add-ons

Mention:

HNSW indexing
Metadata filtering
Multi-query retrieval
Agentic RAG
Caching
Evaluation pipeline
Feedback loop

These keywords make your answer sound like a senior GenAI engineer.

For your GenAI assessment, memorize Questions 21, 22, 25, 27, 32, 38, 41, 44, 49, and 50 especially. Those are the highest-probability advanced RAG 
    questions that interviewers and assessments frequently target.
```
