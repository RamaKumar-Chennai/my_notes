```python
Rama, if you're preparing for a GenAI / RAG Assessment, focus on these areas:

RAG (Retrieval-Augmented Generation) Study Material
1. Why RAG?

LLMs have limitations:

Knowledge cutoff
Hallucinations
No access to private company data

RAG solves this by:

Retrieve relevant documents
Add retrieved context to prompt
LLM generates answer using context

Flow:

User Query
    |
Embedding Model
    |
Vector Search
    |
Relevant Chunks
    |
Prompt + Context
    |
LLM
    |
Final Answer

2. Core Components
Chunking

Large documents → smaller chunks.

Example:

100-page PDF
     ↓
500 chunks


Why?

Embedding models have token limits
Smaller chunks improve retrieval

Types:

Fixed-size chunking
Recursive chunking
Semantic chunking
Embeddings

Convert text to vectors.

Example:

"cat" -> [0.21, 0.55, 0.89]
"dog" -> [0.24, 0.51, 0.86]


Similar meaning → vectors close together.

Popular models:

OpenAI Embeddings
BGE
Sentence Transformers
E5
Vector Database

Stores embeddings.

Examples:

FAISS
Pinecone
Chroma
Weaviate
Milvus

Purpose:

Query Vector
    ↓
Similarity Search
    ↓
Top K Documents

Similarity Search

Metrics:

Cosine Similarity

Most common.

1 = identical
0 = unrelated

Euclidean Distance

Distance between vectors.

Dot Product

Used in some embedding systems.

3. Retrieval
Top-K Retrieval
similarity_search(query, k=5)


Returns five best chunks.

Metadata Filtering

Example:

filter={"year":2024}


Search only selected documents.

Hybrid Search

Combines:

Vector Search
Keyword Search (BM25)

Better than pure vector search.

4. Prompt Construction

Retrieved documents are inserted.

Example:

Use the context below.

Context:
{documents}

Question:
{user_query}


Known as context injection.

5. Reranking

Initial retrieval may return irrelevant chunks.

Use reranker:

Retriever
   ↓
Top 20 docs
   ↓
Reranker
   ↓
Best 5 docs


Improves accuracy.

6. Advanced RAG Concepts
Multi-Query Retrieval

Generate multiple search queries.

Example:

User asks:

What is LangGraph?


System generates:

LangGraph architecture
LangGraph agents
LangGraph workflows


Retrieval becomes stronger.

Self-Query Retrieval

LLM extracts filters automatically.

Example:

Show marketing reports from 2024


RAG creates:

filter={"year":2024,"department":"marketing"}

Parent-Child Retrieval

Store small chunks.

Return larger parent chunks.

Improves context quality.

Contextual Compression

Retrieve many chunks.

Keep only useful sentences.

Reduces token usage.

7. RAG Evaluation Metrics
Precision

Retrieved docs are relevant.

Recall

Relevant docs successfully retrieved.

Faithfulness

Answer supported by context.

Groundedness

Answer comes from documents.

Context Relevance

Retrieved chunks match query.

8. Hallucination Prevention

Methods:

Better chunking
Better retrieval
Reranking
Hybrid search
Citation-based answers
Lower temperature
9. LangChain RAG Pipeline
loader
   ↓
splitter
   ↓
embeddings
   ↓
vector_db
   ↓
retriever
   ↓
prompt
   ↓
LLM


Components you should know:

DocumentLoader
TextSplitter
Embeddings
VectorStore
Retriever
PromptTemplate
Chain

10. Enterprise-Level RAG

Know these terms:

Semantic Search
Hybrid Search
Query Rewriting
Reranking
Agentic RAG
Graph RAG
Multimodal RAG
Knowledge Base
Vector Index
Mock Test (50 Questions)
Beginner (1-20)
1.

What does RAG stand for?

A) Retrieval Augmented Generation ✅

B) Retrieval Assisted Graph

C) Recursive AI Generation

D) Relevant Agent Generator

2.

Which component converts text into vectors?

A) Retriever

B) Embedding Model ✅

C) LLM

D) Parser

3.

Which database is commonly used for vector storage?

A) MySQL

B) Oracle

C) FAISS ✅

D) MongoDB

4.

What is chunking?

A) Compressing data

B) Splitting text into smaller pieces ✅

C) Encrypting text

D) Summarizing document

5.

Why do we chunk documents?

A) Reduce retrieval quality

B) Increase latency

C) Improve retrieval accuracy ✅

D) Remove metadata

6.

What does Top-K refer to?

A) Number of documents retrieved ✅

B) Number of embeddings

C) Number of tokens

D) Chunk size

7.

Which similarity metric is most common?

A) Manhattan

B) Hamming

C) Cosine Similarity ✅

D) Pearson

8.

Which step happens first?

A) Generation

B) Retrieval ✅

C) Fine-tuning

D) Translation

9.

What is a vector?

A) Database table

B) Numerical representation of text ✅

C) API

D) Prompt

10.

Which reduces hallucination?

A) RAG ✅

B) More emojis

C) More RAM

D) HTML

11.

FAISS was developed by?

A) Microsoft

B) Meta ✅

C) Google

D) OpenAI

12.

Embedding models capture:

A) Meaning ✅

B) Font size

C) File name

D) Table count

13.

Document metadata is used for:

A) Filtering ✅

B) Compression

C) Encryption

D) Tokenization

14.

Retriever returns:

A) Context Chunks ✅

B) SQL Tables

C) GPU Memory

D) Tokens

15.

What is context injection?

A) Adding retrieved text into prompt ✅

B) Training model

C) Compression

D) Fine tuning

16.

What is a knowledge base?

A) Collection of documents ✅

B) Prompt only

C) Model weights

D) Dataset labels

17.

What is semantic search?

A) Search based on meaning ✅

B) Search by filename

C) Search by date

D) Search by length

18.

RAG mainly improves:

A) Groundedness ✅

B) GPU speed

C) IDE support

D) Monitor quality

19.

Reranking is performed:

A) After retrieval ✅

B) Before retrieval

C) During embedding

D) During training

20.

Hybrid Search combines:

A) SQL + NoSQL

B) Vector + Keyword ✅

C) Python + Java

D) APIs + GPUs


Frequently Asked Interview Questions
Explain end-to-end RAG architecture.
Why embeddings are needed?
FAISS vs Pinecone?
What is chunk overlap?
What is reranking?
What is hybrid retrieval?
How does RAG reduce hallucinations?
What is Agentic RAG?
What is GraphRAG?
How would you evaluate a RAG system?

For a GenAI assessment, if you master these 50 questions plus concepts like embeddings, chunking, FAISS, retrieval, reranking, hybrid search, 
Agentic RAG, GraphRAG, and evaluation metrics, you'll be well prepared for most 
intermediate-to-advanced RAG tests.
```
