````python
# --- Install dependencies ---
!pip install gradio TTS git+https://github.com/suno-ai/bark.git scipy groq

import gradio as gr
from TTS.api import TTS
from bark import generate_audio
from scipy.io.wavfile import write
from groq import Groq
import os

# --- Setup Groq client ---
client = Groq(api_key="YOUR_GROQ_API_KEY")

# --- Initialize Coqui once ---
coqui_tts = TTS(model_name="tts_models/en/ljspeech/tacotron2-DDC", gpu=True)

# --- Functions ---
def get_commentary(ball_event):
    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[{"role":"user","content":f"Cricket commentary: {ball_event}"}],
        temperature=0.8
    )
    return response.choices[0].message.content

def generate_audio_files(ball_event):
    commentary = get_commentary(ball_event)

    # Coqui output
    coqui_file = "coqui_output.wav"
    coqui_tts.tts_to_file(text=commentary, file_path=coqui_file)

    # Bark output
    bark_file = "bark_output.wav"
    audio_array = generate_audio(commentary)
    write(bark_file, 24000, audio_array)

    return commentary, coqui_file, bark_file

# --- Gradio UI ---
with gr.Blocks() as demo:
    gr.Markdown("## 🏏 Cricket Commentary Generator (Groq + Coqui + Bark)")

    ball_event = gr.Textbox(label="Ball Event", placeholder="e.g., Virat Kohli smashes a six")
    commentary = gr.Textbox(label="Generated Commentary")
    coqui_audio = gr.Audio(label="Coqui TTS", type="filepath")
    bark_audio = gr.Audio(label="Bark TTS", type="filepath")

    btn = gr.Button("Generate Commentary + Audio")

    btn.click(fn=generate_audio_files, inputs=ball_event, outputs=[commentary, coqui_audio, bark_audio])

demo.launch()
````
