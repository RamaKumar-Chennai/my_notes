````python
`num_return_sequences` is the parameter that controls **how many different completions (outputs)** the generator should produce for the same prompt.  

---

## 🔎 How it works
- When you call `generator(prompt, num_return_sequences=1)`, you get **one generated text**.  
- If you set `num_return_sequences=3`, the model will generate **three different variations** of the output, each sampled independently.  
- This is especially useful when you enable **sampling** (`do_sample=True`) with parameters like `temperature`, `top_k`, or `top_p`, because each run can produce a different creative continuation.  

---

## 📌 Example

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

outputs = generator(
    "Once upon a time",
    max_length=30,
    num_return_sequences=3,
    temperature=0.8,
    do_sample=True
)

for i, seq in enumerate(outputs):
    print(f"Sequence {i+1}: {seq['generated_text']}")
```

### Possible Output
```
Sequence 1: Once upon a time there lived a wise old owl...
Sequence 2: Once upon a time in a distant galaxy far away...
Sequence 3: Once upon a time a young boy dreamed of becoming a hero...
```

---

## ⚡ Key Point
- If `do_sample=False` (greedy decoding), multiple sequences may look identical.  
- If `do_sample=True` with randomness (`temperature`, `top_k`, `top_p`), each sequence will likely be different.  

---

👉 In recruiter-style terms:  
- **num_return_sequences = 1** → “Give me one answer.”  
- **num_return_sequences = N** → “Give me N alternative answers to the same question.”  

Would you like me to prepare a **quick MCQ set** contrasting `num_return_sequences`, `temperature`, and `top_p` — so you can instantly recall which knob controls *quantity* vs *creativity* vs *probability filtering*?
````
