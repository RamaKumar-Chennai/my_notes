```python
For this question:

Which command is used to create a Docker image from a Dockerfile?

✅ Answer: docker build

Because docker build reads the Dockerfile and creates a Docker Image.

Example:

docker build -t myapp .

-t myapp → image name
. → location of Dockerfile
Docker Concepts First
Dockerfile

Recipe / Blueprint

FROM python:3.11

WORKDIR /app

COPY . .

RUN pip install -r requirements.txt

CMD ["python", "app.py"]

Docker Image

A packaged application.

Example:

myapp:latest


Think of Image = Class

Docker Container

Running instance of an image.

Think:

Image → Template
Container → Running Application

Important Commands
1. docker build

Creates image from Dockerfile.

docker build -t myapp .


Result:

Dockerfile
    ↓
docker build
    ↓
Docker Image

2. docker run

Creates and starts a new container from image.

docker run myapp


Or

docker run -p 5000:5000 myapp


Result:

Image
   ↓
docker run
   ↓
Container Running

Interview Question

Difference between build and run?

docker build

Creates image

docker run

Runs image as container
3. docker start

Starts an existing stopped container.

Suppose:

docker run --name app1 myapp


Then:

docker stop app1


Later:

docker start app1


✅ Restart existing container

Does NOT create container.

4. docker stop

Stops running container.

docker stop app1

5. docker restart

Restart container.

docker restart app1

6. docker ps

See running containers.

docker ps


Example:

CONTAINER ID
abc123

7. docker ps -a

See all containers.

docker ps -a


Includes:

Running
Exited
Stopped
8. docker images

List images.

docker images


Example:

myapp
python
nginx

9. docker rm

Delete container.

docker rm app1

10. docker rmi

Delete image.

docker rmi myapp

11. docker logs

View logs.

docker logs app1


Very important for debugging.

12. docker exec

Run command inside container.

docker exec -it app1 bash


Enter container shell.

Example:

docker exec -it app1 python


Used frequently in production.

13. docker pull

Download image from Docker Hub.

docker pull nginx


Result:

Docker Hub
    ↓
Local Machine

14. docker push

Upload image to registry.

docker push myrepo/myapp


Useful in CI/CD.

Docker Compose
docker compose

Used to run multiple containers together.

Example:

services:

  app:
    build: .

  db:
    image: postgres


Run:

docker compose up


Result:

App Container
+
Database Container


started together.

Why Compose?

Without compose:

docker run app
docker run postgres
docker run redis


With compose:

docker compose up


Single command.

Commands Commonly Asked in MLOps Assessments
Build image
docker build -t ml-app .

Run container
docker run -p 8000:8000 ml-app

View running containers
docker ps

View logs
docker logs container_id

Enter running container
docker exec -it container_id bash

Stop container
docker stop container_id

Start container
docker start container_id

Multi-container deployment
docker compose up

Easy way to remember
Dockerfile
     ↓
docker build
     ↓
Image
     ↓
docker run
     ↓
Container
     ↓
docker stop/start


For MLOps interviews and assessments, these are the most frequently asked Docker commands along with docker build, docker run, docker ps, docker logs, docker exec, and docker compose.
```
