```python
✅ Objects That Have __dict__
These objects typically store their attributes in a __dict__:
1. User-defined classes
class MyClass: def __init__(self): self.a = 1 self.b = 2 obj = MyClass() print(obj.__dict__) # {'a': 1, 'b': 2} 
2. Modules
import math print(math.__dict__) # Shows all module-level attributes 
3. Functions
def foo(): pass print(foo.__dict__) # Usually empty unless you assign attributes 
4. Classes themselves
print(MyClass.__dict__) # Shows class-level attributes and methods 

❌ Objects That Do Not Have __dict__
These are typically built-in types or optimized objects:
1. Built-in types
•	int, float, str, list, tuple, set, frozenset, bool, etc.
x = 5 print(x.__dict__) # AttributeError 
2. Objects using __slots__
If a class defines __slots__, it may not have a __dict__ unless explicitly included:
class Slim: __slots__ = ['a'] def __init__(self): self.a = 1 obj = Slim() print(obj.__dict__) # AttributeError 
🧠 Why This Matters
•	__dict__ is key for introspection, serialization, and dynamic attribute manipulation.
•	If you're building custom classes for semantic search or embedding wrappers, knowing whether __dict__ is available helps you decide how to store metadata or expose internal state.
Want to explore how __slots__ can optimize memory usage or how __dict__ interacts with inheritance and metaclasses?
Does a dictionary object have __dict__?
No — a regular dictionary object (dict) does not have a __dict__ attribute.
•	__dict__ is used to store attributes of user-defined objects.
•	A dictionary is already a mapping of keys to values — it doesn’t need a separate attribute store. ATTRIBUTE ERROR































lst.reverse()  returns None,it means it changes the original list.
Only list has reverse()

List slicing avoids index error


lst.pop(-3) will return the delted element and chage the original list

b = a (changes wil be reflected here)
c = a.copy() (here no changes reflected if a is changed)

a = []
a.append([1, [2, 3], 4])
a.extend([7, 8, 9])
print(a[0][1][1] + a[2])



a = [x for x in (x for x in 'Geeks 22966 for Geeks' if x.isdigit()) if
(x in ([x for x in range(20)]))] 
print(a)

CODE FOR MATCH OF INTEGER AND STRING
s = 'Geeks 22966 for Geeks'
digits = [x for x in s if x.isdigit()]  # ['2', '2', '9', '6', '6']
matched = [int(x) for x in digits if int(x) in range(10)]






Cleaner Version (if you meant to repeat li 3 times)
li = [2, 3, 9] li = [li for _ in range(3)] print(li) 
Output:
[[2, 3, 9], [2, 3, 9], [2, 3, 9]] 

li = [2, 3, 9]
li = [[x for x in [li]] for _ in range(3)]
print(li)

ACTUAL QUESTION
li = [2, 3, 9] 
li = [[x for x in[li]] for x in range(3)] 
print (li) 


li = ['a', 'b', 'c'] * -3
print(li)


a = ['python', 'learning', '@', 'Geeks', 'for', 'Geeks']  LIST SLICE INDEXING ..THIRD ONE IS INCREMENT
print(a[0:6:2])


a = [1998, 2002] 
b = [2014, 2016] 

print ((a + b)*2)
LISTS ARE JOINED AND THEN REPEATED TWICE



count = 0
for c in (a, b, c):  # Here, 'c' is reused as the loop variable        
{loop var c points to a ,b and c in sep iterations}
    if c[0] == 'Code':
        count += 1
    if c[1] == 'Mcq':
        count += 10

Which of the following is true about lists in Python?
•	A
Lists are dynamic sized arrays
•	B
Lists store references at items contiguous locations
•	C
A list can contain elements of different types.
•	D
All of the above



STRINGS

What are strings in Python?
 
Arrays of bytes representing Unicode characters



How can you print a string without resolving escape sequences in Python?
 
By using the repr() function
What is the primary reason that updating or deleting characters from a string in Python is not allowed?

Python strings are immutable data types

text = "Geeks for Geeks"
index = text.find("for")
print(index)  # Output: 6


text = "Hello, World!"
result = text.startswith("Hello")
print(result)  # Output: True


The title() method in Python converts the first character of each word in a string to uppercase and the rest of the characters to lowercase.
Example:
text = "hello world"
result = text.title()
print(result)           # Output: "Hello World"


text = "hello world, hello Python"
result = text.count("hello")
print(result)            # Output: 2


text = "Hello\tWorld"
print(text.expandtabs(4))




import re

pattern = re.compile(r"\d{3}-\d{2}-\d{4}")
result = pattern.match("Hello World")

print(result)  # Output: None



table = str.maketrans("", "", "aeiou")  # Remove vowels
text = "semantic search"
print(text.translate(table))  # Output: smntc srch


table = str.maketrans("abc", "123")  # Replace 'a'→'1', 'b'→'2', 'c'→'3'
text = "abcde"
print(text.translate(table))  # Output: 123de


d = dict()
for x in enumerate(range(2)):
    d[x[0]] = x[1]
    d[x[1]+7] = x[0]
print(d)


Step-by-Step Execution
enumerate(range(2))
This gives:
[(0, 0), (1, 1)] 
Each x is a tuple: (index, value)
d1 = {'GFG' : 1, 
			'Google' : 2, 
			'GFG' : 3 
print(d1['GFG']);
			} 

keys must be unique in dict…if not,the last key wins..so answer is 3 here.



dict ={} 
print (all(dict))



a = {} 
a.fromkeys(['a', 'b', 'c', 'd'], 98) 
print (a)

Explanation
fromkeys() method returns a new dictionary but does not modify the original dictionary a, so print(a) outputs { } (empty dictionary).

a = {'old': 1}
a.update(dict.fromkeys(['x', 'y', 'z'], 100))
print(a)

in Python 3.9 and above:
You can use the + operator to merge dictionaries!

You can merge lists,tuples and dict using plus operator



SETS
Duplicate values are not allowed in sets..ALSO UNORDERED
print(sorted(set1))  # Output: [1, 2, 3, 4]


set1 = {3, 4, 5}
set1.update({1, 2, 3})  # Set is iterable
print(set1)  # Output: {1, 2, 3, 4, 5}
we can use tuple,list,another set here.but if u add dictionary,it will take only the keys..but for  values ,use dict.values()
set1 = {3, 4, 5}
set1.update({'a': 1, 'b': 2})  # Only keys are added!
print(set1)  # Output: {3, 4, 5, 'a', 'b'}
set1.update({'a': 1, 'b': 2}.values())  # Adds 1 and 2 Only values are added


set1 = {1, 2, 3}        # Original set
set2 = set1.copy()      # Creates a shallow copy of set1
set2.add(4)             # Adds 4 to the copy
print(set1)             # Prints the original set
set1=set2(changes will be reflected)
set.add()
set1 = {1, 2, 3}
set2 = set1.add(4)
print(set2)
set2 is None.
Problem:
•	set.add(4) modifies set1 in place.
•	But it returns None, not the updated set.
•	So set2 becomes None.
```
