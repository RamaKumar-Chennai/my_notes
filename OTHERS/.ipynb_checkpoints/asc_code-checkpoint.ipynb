```python
n=int(input())
lst=[]
res_lst=[]

lst=input().split()

lst=list(map(int,lst))
print(lst)

j=0
while j<n:
  print("j value is ",j)
  print("lst[j] value is ",lst[j])
    
  for i in range(1,(max(lst))):
    temp=(i**3)+(i+1)**3
    if lst[j]==temp:
      print("inside if block")
      res_lst.append(lst[j])
  j+=1
print(res_lst)
res_lst=sorted(res_lst)
print(*res_lst)

```
