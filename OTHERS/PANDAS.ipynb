```python
import pandas as pd

# Original DataFrame
df = pd.DataFrame({
    'A': [1, None, 3],
    'B': [None, 5, 6]
}, index=['x', 'y', 'z'])
print("Original:\n", df)

# ✅ Output:
#    A    B
# x  1  NaN
# y NaN  5.0
# z  3  6.0

# 🔧 Using transform()
transformed = df.transform(lambda col: col.fillna(col.mean()))
print("\nUsing transform():\n", transformed)

# 🖨️ Output of transform():
#      A    B
# x  1.0  5.5
# y  2.0  5.0
# z  3.0  6.0
# - Operates column-wise
# - Fills NaN with the mean of each column
# - Preserves shape exactly

# 🔧 Using apply() (column-wise)
applied = df.apply(lambda col: col.fillna(col.mean()))
print("\nUsing apply() column-wise:\n", applied)

# 🖨️ Output of apply():
#      A    B
# x  1.0  5.5
# y  2.0  5.0
# z  3.0  6.0
# - Also operates column-wise by default (axis=0)
# - Same result as transform() in this case

# 🔧 Using apply() (row-wise)
row_applied = df.apply(lambda row: row.fillna(row.mean()), axis=1)
print("\nUsing apply() row-wise:\n", row_applied)

# 🖨️ Output of apply() with axis=1:
#      A    B
# x  1.0  1.0   # NaN in B replaced with row mean (1.0)
# y  5.0  5.0   # NaN in A replaced with row mean (5.0)
# z  3.0  6.0   # No NaN, row unchanged
# - Operates row-wise
# - Fills NaN with the mean of each row
# - Very different logic from column-wise

# 🧠 Summary
# | Method        | Scope        | Output shape | Typical use |
# |---------------|--------------|--------------|-------------|
# | transform()   | Column-wise  | Same shape   | Element-wise transformations |
# | apply(axis=0) | Column-wise  | Flexible     | Similar to transform() here |
# | apply(axis=1) | Row-wise     | Flexible     | Row-based logic, different results |
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, np.nan, 3],
    'B': [4, 5, np.nan]
})

print("Original:\n", df)

# --- apply() on one column ---
# Apply function to a single column
df['A'] = df['A'].apply(lambda x: x**2 if pd.notna(x) else x)
print("\napply() on one column:\n", df)

# --- apply() on entire DataFrame ---
# Apply function column-wise
df_apply = df.apply(lambda col: col.fillna(col.mean()))
print("\napply() on entire DataFrame:\n", df_apply)

# --- transform() on entire DataFrame ---
# Must return same shape; fills NaN with column mean
df_transform = df.transform(lambda col: col.fillna(col.mean()))
print("\ntransform() on entire DataFrame:\n", df_transform)

# --- transform() on one column ---
df['B'] = df['B'].transform(lambda x: x.fillna(x.mean()))
print("\ntransform() on one column:\n", df)
```

```python
import pandas as pd

s = pd.Series([1, 2, 3, 4, 5])

print("Original Series:\n", s)

# --- expanding() ---
# Expanding mean: cumulative mean up to each row
print("\nExpanding mean:\n", s.expanding().mean())
# Output:
# 0    1.000000   # mean of [1]
# 1    1.500000   # mean of [1,2]
# 2    2.000000   # mean of [1,2,3]
# 3    2.500000   # mean of [1,2,3,4]
# 4    3.000000   # mean of [1,2,3,4,5]

# --- rolling(window=3) ---
# Rolling mean: mean of last 3 consecutive values
print("\nRolling mean (window=3):\n", s.rolling(window=3).mean())
# Output:
# 0    NaN        # not enough values yet
# 1    NaN        # not enough values yet
# 2    2.000000   # mean of [1,2,3]
# 3    3.000000   # mean of [2,3,4]
# 4    4.000000   # mean of [3,4,5]
```

```python
### Key Differences Between Covariance and Correlation

| Aspect        | Covariance                                                                 | Correlation                                                                 |
|---------------|-----------------------------------------------------------------------------|------------------------------------------------------------------------------|
| Definition    | Indicates how two variables vary together (positive, negative, or none).   | Indicates both the direction and strength of the linear relationship.        |
| Range         | Can take any value (unbounded).                                            | Always lies between -1 and +1.                                               |
| Units         | Depends on the units of the variables (e.g., cm·kg).                       | Unitless (standardized measure).                                             |
| Interpretation| Positive → variables increase together; Negative → one increases while the other decreases. | +1 → perfect positive relation; -1 → perfect negative relation; 0 → no linear relation. |
| Formula       | \(\mathrm{Cov}(X,Y)=\frac{\sum (X_i-\bar{X})(Y_i-\bar{Y})}{n-1}\)          | \(\mathrm{Corr}(X,Y)=\frac{\mathrm{Cov}(X,Y)}{\sigma_X \cdot \sigma_Y}\)     |
| Use Case      | Preliminary measure of relationship direction.                             | Standardized measure, useful for comparing across datasets.                  |
```

```python
 Correlation and Linear Relationships
- Correlation (Pearson’s correlation) specifically measures the strength of a linear relationship between two variables.
- If there is no linear relationship, the correlation coefficient will be close to 0.
- But 0 correlation does not mean “no relationship at all” — it only means no linear relationship. There could still be a non-linear relationship (curved, quadratic, exponential, etc.) that correlation doesn’t capture.

📘 Example
- Suppose X is a variable and Y=X^2.
- The relationship is perfectly deterministic (non-linear).
- But Pearson correlation between X and Y can be 0, because the upward and downward trends cancel out symmetrically.
- So correlation misses this non-linear dependency.

✅ Key Takeaway
- No linear relationship → correlation ≈ 0.
- But correlation = 0 does not imply independence or “no relationship.” It only rules out linear dependence.
- To detect non-linear relationships, you’d use other methods (e.g., Spearman’s rank correlation, mutual information, scatter plots).
```

```python
Height and Weight
- Height and weight often show positive correlation (taller people tend to weigh more).
- But they are not the same variable — they carry different information.
- Dropping one just because they’re correlated is risky:
- In ML, correlated features can cause multicollinearity (especially in linear models), but they may still add predictive power in non-linear models (like decision trees, random forests, neural nets).
- Instead of blindly dropping, you can use feature selection or dimensionality reduction (e.g., PCA) to handle correlated features.

2. Height in cm vs Height in m
- This is a duplicate representation of the same variable.
- One is just a scaled version of the other (1\mathrm{\  m}=100\mathrm{\  cm}).
- Keeping both adds no new information and can confuse the model.
- ✅ In this case, you should drop one column.

🔑 Rule of Thumb
- Highly correlated but different variables (height vs weight): Don’t drop automatically; check feature importance or use dimensionality reduction.
- Exact duplicates or scaled versions (height in cm vs m): Always drop one, since they’re redundant.
```

```python
Height in cm vs Height in m
- Height in cm and Height in m are perfectly linearly related:
\mathrm{Height\  in\  cm}=100\times \mathrm{Height\  in\  m}- This is not just collinearity, it’s perfect multicollinearity (one variable is an exact linear transformation of the other).
- In regression or ML models:
- Perfect multicollinearity means one column can be predicted exactly from the other.
- Algorithms like linear regression will fail or give unstable coefficients if both are included.
- Tree-based models or neural nets won’t “fail,” but the duplicate feature adds no new information and can confuse feature importance.

✅ What to Do
- Drop one column (either cm or m). Keeping both is redundant.
- General rule: if two features are exact duplicates or scaled versions, remove one to avoid perfect multicollinearity.

⚖️ Difference from “high correlation”
- High correlation (e.g., height vs weight): Variables are related but not exact duplicates. You may keep both, depending on the model.
- Perfect collinearity (e.g., height in cm vs m): One is a deterministic transformation of the other. You must drop one.

👉 So yes, height in cm vs m is collinearity — specifically perfect multicollinearity.
Would you like me to show you a Python snippet that detects perfect multicollinearity in a dataset (e.g., using variance inflation factor or correlation checks)? That way you can automate dropping redundant columns before modeling.
```

```python
Covariance
- Covariance measures whether two variables move together (positive or negative).
- It does not restrict itself to linearity — any kind of dependency (linear or non-linear) can influence covariance.
- But here’s the catch: covariance is an average measure across the dataset.
- If part of the data shows positive co‑movement and another part shows negative co‑movement, they can cancel out.
- Example: Y=X^2 with symmetric X around 0 → covariance ≈ 0, even though the relationship is strong and non-linear.

🔎 Correlation (Pearson)
- Correlation is just standardized covariance.
- By dividing covariance by the product of standard deviations, it focuses only on linear strength.
- So correlation = 0 means “no linear relationship,” but doesn’t rule out non-linear ones.

✅ Key Insight
- Covariance can reflect non-linear relationships, but only if the overall trend doesn’t cancel out.
- Correlation cannot capture non-linear relationships at all — it’s strictly about linearity.
- Both measures can miss strong non-linear dependencies if the dataset is symmetric.

📘 Practical ML Tip
- To detect non-linear relationships, you need other tools:
- Spearman’s rank correlation → captures monotonic (always increasing/decreasing) relationships, even if curved.
- Mutual information → captures any dependency, linear or non-linear.
- Scatter plots → always a good sanity check.
```

```python
 What Zero Covariance Means
- Covariance = 0 → On average, the variables do not move together in a consistent way.
- But this does not mean “no relationship at all.”
- It only means there is no overall linear co‑movement.
- Non-linear relationships can still exist (e.g., Y=X^2 with symmetric X around 0 → covariance = 0, but the relationship is perfectly deterministic).

🔎 What Zero Correlation Means
- Correlation = 0 → No linear relationship (since correlation is just standardized covariance).
- Again, this does not rule out non-linear relationships.
- Example: Y=X^2 → correlation = 0, but clearly Y depends on X.

✅ Key Insight
- Zero covariance and zero correlation together →
- Means there is no linear relationship.
- But it does not guarantee independence or “no relationship at all.”
- Strong non-linear dependencies can still exist but remain invisible to these measures.

📘 Practical ML Lesson
- If you see zero correlation/covariance, don’t assume the variables are unrelated.
- Always check with:
- Scatter plots (visual check for non-linear patterns).
- Spearman’s rank correlation (captures monotonic non-linear trends).
- Mutual information (captures any dependency, linear or non-linear).

👉 So the short answer:
- Zero covariance/correlation = no linear relationship.
- It does not rule out non-linear relationships.
- Only if you also check independence (via probability theory or mutual information) can you say “no relationship at all.”
```

```python
df.sample()
is a pandas method used to randomly select rows from a DataFrame — perfect for quick previews, testing, or creating randomized subsets of your data.

🔧 Syntax
df.sample(n=5, frac=None, replace=False, random_state=None)


Key Parameters:
- n: Number of rows to return
- frac: Fraction of rows to return (e.g., frac=0.1 for 10%)
- replace: If True, allows sampling with replacement
- random_state: Seed for reproducibility
```

```python
What random_state Really Is
- In Pandas (df.sample) and scikit-learn (train_test_split, etc.), random_state is just a seed number for the random number generator.
- It ensures reproducibility: if you set the same random_state, you’ll always get the same random sample or split.

📘 Range of random_state
- It can be any integer (positive, negative, or zero).
- Common practice: use small positive integers (like 0, 1, 42) — but technically, any integer works.
- There’s no concept of “low” or “high” meaning more or less randomness.
- random_state=1 and random_state=9999 are equally valid — they just produce different deterministic sequences.

✅ Key Takeaway
- Low vs high doesn’t matter — it’s not a scale of randomness.
- It’s simply a “label” for the random seed.
- If you want reproducibility, pick a fixed integer (e.g., random_state=42).
- If you don’t care about reproducibility, leave it as None (default), and you’ll get different samples each run.

👉 Would you like me to show you a quick demo in Python where I sample with different random_state values, so you can see how the rows change but remain consistent for the same seed?
```

```python
s a pandas method that shows you how much memory each column in your DataFrame consumes, measured in bytes. It's super handy when you're working with large datasets and want to optimize performance or reduce memory footprint.

🔧 Syntax
df.memory_usage(index=True, deep=False)


Parameters:
- index=True: includes memory used by the index
- deep=True: gives a more accurate estimate, especially for object/string columns

🧪 Example
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': ['apple', 'banana', 'cherry']
})

print(df.memory_usage())
print(df.memory_usage(deep=True))


✅ Output (approximate):
Without deep=True:
Index     128
A          24
B          24
dtype: int64


With deep=True:
Index     128
A          24
B         152
dtype: int64


- 'B' is an object column (strings), so deep=True reveals its true memory usage.
- 'A' is numeric, so its memory is straightforward.
```

```python
- Your column df["Date Of Stop"] has dtype 'O' (object). In pandas, an object dtype often means the values are strings, though technically it can hold mixed Python objects.
- The .str accessor in pandas is designed to work on Series of strings. If the column is object dtype and contains strings, .str.contains() will work directly.
- If the column has non-string values (like NaN, integers, or other Python objects), then .str.contains() may raise an error. In that case, you’d need to convert explicitly with .astype(str) or handle missing values.
So your code works fine as long as all entries are strings (or NaN, which .str.contains() can handle if you pass na=False).
Example:
dash_mask = df["Date Of Stop"].str.contains("-", na=False)
slash_mask = df["Date Of Stop"].str.contains("/", na=False)


If you suspect mixed types in the column, you can enforce string conversion safely:
dash_mask = df["Date Of Stop"].astype(str).str.contains("-", na=False)
slash_mask = df["Date Of Stop"].astype(str).str.contains("/", na=False)


👉 In short: no need to convert if the column is already strings, but explicit conversion is a defensive move when you’re unsure about mixed types.
```

```python
Ah, now that’s a different situation. If your column shows dtype('<M8[ns]'), that means pandas has already parsed it as a datetime64[ns] type, not as strings.
Here’s the key point:
- The .str accessor only works on string/object dtype columns.
- Since your column is datetime64, you can’t directly use .str.contains() — it will raise an error.
- Instead, you should either convert the datetime column back to string, or use datetime-specific accessors.
Option 1: Convert to string for pattern checks
dash_mask = df["Date Of Stop"].astype(str).str.contains("-", na=False)
slash_mask = df["Date Of Stop"].astype(str).str.contains("/", na=False)


Option 2: Use datetime accessors
If you’re checking for formatting (like dash vs slash), that’s really about the original raw data before parsing. Once pandas has converted to datetime64, the internal representation doesn’t preserve whether the input had dashes or slashes — it’s just a datetime object.
So if your goal is to detect whether the raw data had "-" or "/", you should do the .str.contains() before converting to datetime. After conversion, that information is lost.

👉 In short:
- With dtype('O'), .str.contains() works directly.
- With dtype('<M8[ns]'), you must convert to string first, but you’ll only see the standardized datetime format (usually with dashes). The original slash vs dash distinction is gone.
```
