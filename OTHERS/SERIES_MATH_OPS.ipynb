```python
import pandas as pd

s = pd.Series([10, 20, 30])
s[1] = 99  # Changing value at index 1

print(s)
```

```python
s[3]=200
print(s)
```

```python
s = s.drop(2)
print(s)
```

```python
s=s.reset_index()
```

```python
print(s)
```
