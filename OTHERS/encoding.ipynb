```python
In machine learning, encoding is the process of converting categorical data into numerical formats that algorithms can process. Here are the common types of encoding techniques:

1. Label Encoding
Converts each category into a unique integer.
Example: ["Red", "Green", "Blue"] → [0, 1, 2].
Suitable for ordinal data but may introduce unintended relationships for nominal data.
2. One-Hot Encoding
Creates binary columns for each category.
Example: ["Red", "Green", "Blue"] → [[1, 0, 0], [0, 1, 0], [0, 0, 1]].
Works well for nominal data but can increase dimensionality.
3. Ordinal Encoding
Assigns integers to categories based on their order.
Example: ["Low", "Medium", "High"] → [0, 1, 2].
Best for ordinal data where order matters.
4. Binary Encoding
Converts categories into binary numbers and splits them into columns.
Example: ["A", "B", "C"] → [01, 10, 11].
Reduces dimensionality compared to one-hot encoding.
5. Frequency Encoding
Replaces categories with their frequency in the dataset.
Example: ["A", "A", "B", "C"] → [2, 2, 1, 1].
Useful for large datasets with many categories.
6. Target Encoding
Replaces categories with the mean of the target variable for each category.
Example: If ["A", "B", "C"] corresponds to target values [0.5, 0.7, 0.2], it encodes as [0.5, 0.7, 0.2].
Commonly used in supervised learning but may lead to data leakage.
7. Hash Encoding
Maps categories to a fixed number of columns using a hash function.
Example: ["A", "B", "C"] → [hash(A), hash(B), hash(C)].
Useful for high-cardinality data.

Each technique has its strengths and is chosen based on the type of data (nominal or ordinal) and the specific requirements of the machine learning model.
```
