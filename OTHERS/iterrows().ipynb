```python
In pandas, iterrows() is a method used to iterate over rows of a DataFrame as (index, Series) pairs. While it can be useful in certain scenarios, it is generally not the most efficient way to process data in pandas due to its row-by-row nature. Vectorized operations are usually preferred for performance.

Here’s an example of how to use iterrows():

Example 1: Basic Usage
Copy the code
import pandas as pd

# Sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['Pune', 'Mumbai', 'Delhi']}
df = pd.DataFrame(data)

# Iterating through rows
for index, row in df.iterrows():
    print(f"Index: {index}, Name: {row['Name']}, Age: {row['Age']}, City: {row['City']}")

Example 2: Modifying Data (Not Recommended)
Copy the code
# Adding a new column based on iteration
for index, row in df.iterrows():
    df.loc[index, 'Age Group'] = 'Adult' if row['Age'] >= 18 else 'Minor'

print(df)

Notes:
Performance: iterrows() is slow for large DataFrames. Use vectorized operations or apply() for better performance.
Data Type: The row returned by iterrows() is a pandas Series, which may not preserve data types as expected.

For large-scale operations, consider alternatives like apply(), itertuples(), or vectorized operations for efficiency.
```
