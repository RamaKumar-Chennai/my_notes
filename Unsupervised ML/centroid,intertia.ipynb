```python
Not quite — inertia isn’t the “mean of means.” Let me clarify:

- **Within each cluster**:  
  You compute the centroid (mean of the points assigned to that cluster).  

- **Inertia definition**:  
  For every point in the dataset, you calculate the squared distance between the point and its assigned cluster centroid.  
  Then you **sum all those squared distances across all clusters**.  

\[
\text{Inertia} = \sum_{i=1}^{n} \| x_i - \mu_{c(i)} \|^2
\]

where:  
- \(x_i\) = data point  
- \(\mu_{c(i)}\) = centroid of the cluster that \(x_i\) belongs to  
- \(n\) = total number of points  

So inertia is a **single number** that measures how tightly the clusters fit the data. Lower inertia means points are closer to their centroids, i.e. better clustering.  

👉 Important distinction:  
- You don’t take the mean of the cluster means.  
- You take the **sum of squared distances of all points to their cluster centroid**.  

That’s why KMeans chooses the run with the lowest inertia — it’s the one where points are most compact around their centroids.  

Would you like me to walk through a tiny numeric example (say 6 points split into 2 clusters) to show exactly how inertia is computed step by step?
```
