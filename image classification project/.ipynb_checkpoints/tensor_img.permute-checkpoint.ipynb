```python
import torch
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
from PIL import Image

# Load an image using PIL
image_path = "TIGER.jpg"  # Replace with your image path
original_image = Image.open(image_path)

# Convert the image to a tensor
transform = transforms.ToTensor()
tensor_image = transform(original_image)

# Convert the tensor back to a format suitable for display
# PyTorch tensors are channel-first (C, H, W), but Matplotlib expects (H, W, C)
image_for_display = tensor_image.permute(1, 2, 0).numpy()

# Display the image
plt.imshow(image_for_display)
plt.axis('off')  # Optional: Turn off axis
plt.show()


```
