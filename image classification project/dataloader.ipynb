```python
In PyTorch, the DataLoader is a utility that simplifies and optimizes the process of loading data for training and evaluation. It works with datasets to provide features like batching, shuffling, and parallel data loading.

Key Features of DataLoader:
Batching: Automatically divides the dataset into smaller batches.
Shuffling: Randomizes the order of data to reduce bias during training.
Parallel Loading: Uses multiple worker threads to load data faster.
Basic Syntax:
Copy the code
from torch.utils.data import DataLoader

# Example usage
dataloader = DataLoader(dataset, batch_size=32, shuffle=True, num_workers=4)

Parameters:
dataset: The dataset to load (e.g., torch.utils.data.Dataset).
batch_size: Number of samples per batch (default: 1).
shuffle: Whether to shuffle the data (default: False).
num_workers: Number of subprocesses for data loading (default: 0, meaning data is loaded in the main process).
Example Code:
Copy the code
import torch
from torch.utils.data import DataLoader, Dataset

# Custom dataset
class MyDataset(Dataset):
    def __init__(self):
        self.data = torch.arange(100)  # Example data
        self.labels = torch.arange(100) % 2  # Example labels

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        return self.data[idx], self.labels[idx]

# Create dataset and DataLoader
dataset = MyDataset()
dataloader = DataLoader(dataset, batch_size=10, shuffle=True)

# Iterate through DataLoader
for batch_idx, (data, labels) in enumerate(dataloader):
    print(f"Batch {batch_idx}:")
    print(f"Data: {data}")
    print(f"Labels: {labels}")


This example demonstrates how to create a custom dataset and use the DataLoader to iterate through it in batches. You can adapt it to your specific dataset and requirements.
```

```python
Best Practices
Combine with num_workers
The num_workers parameter in the DataLoader controls the number of subprocesses to use for data loading. Combining pin_memory with a non - zero num_workers can further speed up the data loading process. For example:

dataloader = DataLoader(dataset, batch_size = 10, pin_memory = True, num_workers = 4)

In this case, the data will be loaded in parallel by 4 subprocesses, and the batches will be stored in pinned memory.

```
