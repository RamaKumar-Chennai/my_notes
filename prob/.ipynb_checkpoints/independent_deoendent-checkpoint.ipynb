```python
Here’s a clear set of examples contrasting independent and dependent choices, so you can see the difference in action:

🎯 Independent Choice Examples
These are situations where one choice does not affect the other.
- Meals
- 3 breakfast options, 2 lunch options.
- Total = 3\times 2=6.
- No division needed.
- Clothes
- 4 shirts, 5 pants.
- Total outfits = 4\times 5=20.
- Dice Rolls
- Roll a red die and a blue die.
- Red die: 6 outcomes, Blue die: 6 outcomes.
- Total = 6\times 6=36.

🎲 Dependent Choice Examples
These are situations where one choice does affect the other.
- Drawing Balls Without Replacement
- Bag has 7 balls. Draw 2.
- Ordered outcomes = 7\times 6=42.
- If order doesn’t matter → divide by 2 → 21.
- Selecting Students for a Team
- 10 students, choose 3.
- Ordered outcomes = 10\times 9\times 8=720.
- Unordered (committee) → divide by 3!=6.
- Total = 120.
- Cards from a Deck
- Draw 2 cards from 52 without replacement.
- Ordered outcomes = 52\times 51=2652.
- Unordered → divide by 2 → 1326.

🔑 Rule of Thumb
- Independent choices → multiply directly.
- Dependent choices → multiply, then divide by factorial if order doesn’t matter.
```
