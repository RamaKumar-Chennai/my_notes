```python
import re
text='I have a  pen to draw'
re.findall(
   # r"\ba\b\s+(\w+)\s+\bto\b\s+(\w+)", text 
    r"\ba\b\s+(\w+)", text
)
```

```python
text  = " the  cities are polution made city contaminated, polution incereses due to population po hi"
re.findall(r'\bpo\w+\b', text )
#re.findall(r'po\w+', text )
```

```python
text = "I am satish, my email is satish@gmail.com and this is my another email, test@hotmail.com , apple@apple.co.in this will be the next text "
re.findall(r"[a-zA-Z0-9._]+@+[a-zA-Z]+\.+[a-zA-Z.]{2,}"   ,  text)
```

```python
text1 = "I am satish, my email is satish@gmail.com and this i@@@@s my ....another email"
email=re.findall(r"[a-zA-Z0-9._]+@+[a-zA-Z]+\.+[a-zA-Z.]{2,}"   ,  text1)[0]
print(email)
```

```python
re.sub(r"[a-zA-Z0-9._]+@+[a-zA-Z]+\.+[a-zA-Z.]{2,}"  ,"",text1)
print(text1)
```
