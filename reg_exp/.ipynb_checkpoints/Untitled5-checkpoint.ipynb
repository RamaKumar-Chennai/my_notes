```python
text = "I will play football at 10am 4pm 8pm with my 20 friends. my email is sample@gmail.com"

import re

re.findall(r"\d{1,2}(?:am|pm)", text)
```

```python
text = "I will play football at 10am 4pm 8pm with my 20 friends football i liek football. my email is sample@gmail.com"
re.findall(r"I.* football", text)
```

```python
text  = "I have a pen to draw"
re.findall(
    r"a\s+(\w+)", text 
    
)
```
