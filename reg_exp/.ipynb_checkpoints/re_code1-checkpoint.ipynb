```python
import pandas as pd
data=pd.read_json('data1.json')
data
```

```python
data.info()
```

```python
data['day']=data['Date'].dt.day
data['month']=data['Date'].dt.month
data['year']=data['Date'].dt.year

data
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(15,5))
sns.histplot(data=data,x='MAG',color='red',label='MAGAZINE')        
sns.histplot(data=data,x='NEWS',color='yellow',label='NEWS')

plt.legend()

plt.xlabel('MAG Vs NEWS')
plt.title('Comp. BW News and Mag')
plt.show()
```

```python
# Example Task:

""" I will go to m$$eting at 4pm and will be %% back around 10pm. mean time please completed the work and send to 
arun@hotmail.com and sam_09@gmail.com. once you sent call me on 99725263 or 92663-29273 numbers.



for the above text,

1. remove spl char 
2. get email 
3. get phone numbers
4. total numbers of ermails
5. when he will be c]back from meeting
6. remove all 1 and 2 letter words.
"""
```

```python
text="I will go to m$$eting at 4pm and will be %% back around 10pm. mean time please completed the work and send to arun@hotmail.com and sam_09@gmail.com. once you sent call me on 99725263 or 92663-29273 numbers."
import re

email=re.findall(r"[a-zA-Z0-9._]+@+[A-Za-z]+\.+[a-zA-Z.]{2,}",text)
print("the email ids are")
for i in range(0,len(email)):
    print(email[i])
print("the number of email ids is ",len(email))


ph_no=re.findall(r'\d{5}-\d{5}',text)

ph_no.append(re.findall(r'\d{8,}',text)[0])
print("the phone numbers are ")
for i in range(0,len(ph_no)):
    print(ph_no[i])

print("the number of phone numbers is",len(ph_no))
```

```python
text="I will go to m$$eting at 4pm and will be %% back around 10pm. mean time please completed the work and send to arun@hotmail.com and sam_09@gmail.com. once you sent call me on 99725263 or 92663-29273 numbers."
import re
text1=re.sub(r'[^A-Za-z0-9@._  -]','',text)
print(text1)
time_lst=re.findall(r'back\s+(\w+\s+\d{1,}(?:am|pm))',text)
print("he will be back from meeting  ",*time_lst)
```

```python
import re

text2=re.findall(r'\b\w{1,2}\b',text1)
print(text2)
text3=re.sub(r'\b\w{1,2}\b'," ",text1)
print(text3)
```
