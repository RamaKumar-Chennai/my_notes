```python
import re
text1 = "I am satish, my email is satish@gmail.com and this i@@@@s my ....another email"
re.findall(r"[a-zA-Z0-9._]+@+[a-zA-Z]+\.+[a-zA-Z.]{2,}" ,text1)
text2=re.sub(r"[a-zA-Z0-9._]+@+[a-zA-Z]+\.+[a-zA-Z.]{2,}" ,'',text1)
print(text2)
```

```python
text3=re.sub(r"[^a-zA-Z0-9]"," ",text2)
print(text3)
```
