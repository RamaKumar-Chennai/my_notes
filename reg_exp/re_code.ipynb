```python
import re

txt = "The rain in Spain"
x = re.search(r"\bS\w+", txt)
print(x.string)
print(x)
print(x.group())
print(x.start())
print(x.span())
print(x.end())
print(x.endpos)
print(x.pos)

C:\Users\Rama Kumar\AppData\Local\Programs\Python\Python36-32\Scripts>pip --version
```
