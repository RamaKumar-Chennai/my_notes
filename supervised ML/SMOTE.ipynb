```python
Oversampling vs. SMOTE
- Oversampling = Any technique that increases the number of minority class samples (e.g., duplicating rows, random oversampling, or synthetic generation like SMOTE).
- SMOTE = A specific oversampling method that creates synthetic samples by interpolating between existing minority class points.

📌 Where do we apply oversampling?
- Training set only ✅
- Oversampling is applied only to the training data.
- Reason: If you oversample the test set, you artificially inflate minority samples and leak information about the distribution into evaluation.
- This makes your test set no longer a true reflection of real-world data, leading to data leakage and overly optimistic performance metrics.
- Test set / Validation set ❌
- Must remain untouched, imbalanced as in reality.
- This ensures that your evaluation reflects how the model will perform on unseen, naturally imbalanced data.

⚙️ Workflow Example
- Split dataset first → train_test_split(X, y)
- Apply oversampling (SMOTE or RandomOversampler) only on training set
from imblearn.over_sampling import SMOTE
smote = SMOTE()
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
- Train model on resampled training set
- Evaluate model on untouched test set

🔑 Key Takeaway
- Oversampling (including SMOTE) is always applied only on the training dataset.
- The test set must remain untouched to avoid leakage and preserve realistic evaluation.
```
